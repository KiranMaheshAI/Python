Dynamic Programming (DP)

DP solves complex problem by breaking them into overlapping subproblems, solving each subproblem once, and storing results to avoid redundant computations.

#### 1D Dynamic Proogramming

Fibonacci series

```python
def fibonacci_series(n):
    if n <= 1:
        return n
    fib = [0]*(n+1)
    fib[1] = 1
    for i in range(2, n+1):
        fib[i] = fib[i-1]+fib[i-2]
    return fib[n-1]
fibonacci_series(6)
```

```python
def climbing_stairs(n):
    if n<=2:
        return n
    dp = [0]*(n+1)
    dp[1] = 1
    dp[2] = 2
    for i in range(3, n+1):
        dp[i] = dp[i-1]+dp[i-2]
    return dp[-1]

climbing_stairs(6)
```

### House Robber

```python
# Max money without robbing adjacent bank
def house_robbers(arr):
    n = len(arr)
    dp = [0] * n
    dp[0] = arr[0]
    dp[1] = max(arr[0], arr[1])
    for i in range(2, n):
        dp[i] = max(dp[i-1], arr[i]+ dp[i-2])
    print(dp)
    return dp[-1]

house_robbers([1, 2, 3, 4, 1])
```

### 0/1 Knapsack

```python
# Max value with weight capacity W, each item used at most once
def knapsack(weights, values, W):
    n = len(weights)
    dp = [[0] * (W + 1) for _ in range(n + 1)]

    for i in range(1, n + 1):
        for w in range(W + 1):
            # Don't take item i
            dp[i][w] = dp[i-1][w]
            # Take item i (if it fits)
            if weights[i-1] <= w:
                dp[i][w] = max(dp[i][w],
                               dp[i-1][w - weights[i-1]] + values[i-1])
    print(dp)
    return dp[n][W]

weights=[2,3,10]
values=[3,2,5]
W=5

knapsack(weights, values, W)
```

## Longest Common SubSequence

```python
# Using Recursion
def longest_common_subsequence_helper(str1, str2, i, j):
    if i == len(str1) or j == len(str2):
        return 0
    if str1[i] == str2[j]:
        return 1 + longest_common_subsequence_helper(str1, str2, i+1, j+1)
    else:
        return max(longest_common_subsequence_helper(str1, str2, i+1, j), longest_common_subsequence_helper(str1, str2, i, j+1))
        
longest_common_subsequence_helper("ABCDGH", "AEDFHR", 0, 0)
```

```python
# Using Dynamic Programming
def lcs_dp(s1, s2):
    m, n = len(s1), len(s2)
    dp = [[0]*(n+1) for _ in range(m+1)]
    for i in range(1, m+1):
        for j in range(1, n+1):
            dp[i][j] = max(dp[i-1][j], dp[i][j-1])
            if s1[i-1] == s2[j-1]:
                dp[i][j] = 1+ max(dp[i-1][j], dp[i][j-1])
    return dp

lcs_dp("ABCDGH", "AEDFHR")
```
