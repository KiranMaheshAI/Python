## Binary Search

Given a sorted array, you should find the searching element in 'O(logn)' 

```python
nums = [1, 3, 5, 7, 9, 11, 13]
def binary_search(nums, target):
    n = len(nums)
    if len(nums)<1:
        return nums
    start = 0
    end = len(nums)-1
    while(start <= end):
        mid = (start+end)//2
        print(f"{mid} and {nums[mid]}")
        if target == nums[mid]:
            return mid
        elif target > nums[mid]:
            start = mid+1
        else:
            end = mid-1
    return -1

print(binary_search(nums, 13))
    
```

```python
# Find First True (Left Boundary)
def find_left_boundary(nums, target):
    start, end = 0, len(nums)-1
    result = -1
    while start<=end:
        mid = (start+end)//2
        if nums[mid] == target:
            result = mid
            end = mid - 1
        elif nums[mid] > target:
            end = mid - 1
        else:
            start = mid + 1
    return result

print(find_left_boundary([1, 2, 3, 3, 3, 3, 5, 6, 7, 7, 8 ], 7))
```

```python
# Find Right Boundary
def find_right_boundary(nums, target):
    start, end = 0, len(nums)-1
    result = -1
    while start<=end:
        mid = (start+end)//2
        if nums[mid] == target:
            result = mid
            start = mid + 1
        elif nums[mid] > target:
            end = mid - 1
        else:
            start = mid + 1
    return result

print(find_right_boundary([1, 2, 3, 3, 3, 3, 5, 6, 7, 7, 7, 8 ], 7))
```

##  Search in Rotated Sorted Array

Array was sorted, then rotated at some pivot: [4,5,6,7,0,1,2]. One half is always sorted — use that to decide which half to search.

```python
# nums = [4, 5, 6, 7, 0, 1, 2]
nums1 = [1, 3, 5, 7, 9, 11, 13]
def search_rotated_array(nums, target):
    start, end = 0, len(nums)-1
    while start <= end:
        mid = (start+end)//2

        if nums[mid] == target:
            return mid
        elif target > nums[mid] or nums[mid] < target <= nums[end]:
            if nums[end] >= target:
                start = mid + 1
            else:
                end = mid - 1
        else:
            if nums[start] <= target < nums[mid]:
                end = mid - 1 
            else:
                start = mid + 1
    return -1 
print(search_rotated_array(nums1, 13))
```

## Binary Search on Answer

Find the minimum capacity to ship packages in D days.

```python
def ship_within_days(weights: list[int], days: int) -> int:
    left = max(weights)        # minimum possible capacity
    right = sum(weights)       # maximum possible capacity
    print(f"max weight per day {left} and total weights is {right}")
    def can_ship(capacity):
        needed_days, current = 1, 0
        for w in weights:
            if current + w > capacity:
                needed_days += 1
                current = 0
            current += w
        return needed_days <= days

    while left < right:
        mid = left + (right - left) // 2
        print(f"Left {left}, right {right} and mid {mid}")    
        if can_ship(mid):
            right = mid         # try smaller capacity
        else:
            left = mid + 1      # need more capacity
    print(f"Left {left}, right {right} and mid {mid}")
    return left

print(ship_within_days([1, 2, 4, 5, 6, 7, 8], 4))
```

## Allocate books

Given n books with pages[i] pages each, and m students, allocate contiguous books to each student to minimize the maximum pages any student reads. Each student must get at least one book.

```python
pages = [12, 34, 67, 90]
students = 2

def can_allocate(pages, students, mid):
    current = 0
    students_needed = 1
    for page in pages:
        if current+page > mid:
            students_needed += 1
            current = 0
        current += page
    return students_needed <= students

def allocate_books(pages, students):
    if students > len(pages):
        return -1
    left = max(pages)
    right = sum(pages)
    result = 0
    while left <= right:
        mid  = left + (right-left)//2
        print(f"mid is {mid}")
        if can_allocate(pages, students, mid):
            result = mid
            right = mid - 1
        else:
            left = mid + 1
    return result

allocate_books(pages, students)
```

```python
# Anthropic Answer
def can_allocate(pages: list[int], students: int, max_pages: int) -> bool:
    students_needed = 1
    current_load = 0

    for p in pages:
        if p > max_pages:
            return False            # single book exceeds limit

        if current_load + p > max_pages:
            students_needed += 1    # give this book to next student
            current_load = p

            if students_needed > students:
                return False        # need too many students
        else:
            current_load += p

    return True

def allocate_bookss(pages: list[int], students: int) -> int:
    if students > len(pages):
        return -1                   # impossible — not enough books

    left = max(pages)
    right = sum(pages)
    result = right
    
    while left <= right:
        mid = left + (right - left) // 2
        print(f"mid is {mid}")
        if can_allocate(pages, students, mid):
            result = mid            # valid — try smaller max
            right = mid - 1
        else:
            left = mid + 1          # invalid — need larger max

    return result
allocate_bookss(pages, students)
```

### Aggressive Cows (Maximize Minimum Distance)

Given n stalls at positions stalls[] and c cows, place cows in stalls such that the minimum distance between any two cows is maximized.

Input:  stalls = [1, 2, 4, 8, 9], cows = 3
Output: 3

```python
def can_place(stalls, cows, mid):
    last_placed  = stalls[0]
    count = 1
    for i in range(1, len(stalls)):
        if stalls[i] - last_placed >= mid:
            count += 1
            last_placed = stalls[i]
            if count == cows:
                return True
    return False

def aggressive_cows(stalls, cows):
    start = 1
    end = stalls[-1] - stalls[0]
    result = 0
    while start <= end:
        mid = start + (end-start)//2
        if can_place(stalls, cows, mid):
            result = mid
            start = mid + 1
        else:
            end = mid - 1
    return result

print(aggressive_cows([1, 2, 4, 8, 9], 3))
```

## Median of Two Sorted Arrays

## Binary search in increasing order matrix:
Consider the increasing order matrix with positive values , search for the element in O(m * n) time complexity. 

Ex: matrix = [[1, 3, 5], [7, 9, 10], [12, 14, 16]] and element is 15

```python
def search_matrix(matrix, target):
    if not matrix or not matrix[0]:
        return False

    m = len(matrix)
    n = len(matrix[0])

    row = 0
    col = n - 1  # start from top-right

    while row < m and col >= 0:
        if matrix[row][col] == target:
            return True
        elif matrix[row][col] > target:
            col -= 1
        else:
            row += 1

    return False
```
