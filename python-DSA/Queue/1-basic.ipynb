### Queue

Queues are Linear Data Strcuture where we will follow First In First Out (FIFO)

Queues are Abstract Data type where we can implement using the List and Linked List.

### Circular Queue

Last Node is connected to my first node.

It is also know as Ring Buffer.

### Priority Queue

Elements are served based on the priority.
