## Priority Queue

A priority queue is like a regular queue, but each item has a priority. Instead of being served in the order they arrive, items with higher priority are served first.

Key properties of priority queue:

1. High-priority elements are dequeued before low-priority ones.
2. If two elements have the same priority, they are dequeued in their order of insertion like a queue.

Functions-

1. put() - Puts an item into the queue.
2. get() - Removes and returns an item from the queue.
3. qsize() - Returns the current queue size.
4. empty() - Returns True if the queue is empty, False otherwise. It is equivalent to qsize()==0.
5. full() - Returns True if the queue is full, False otherwise. It is equivalent to qsize()>=maxsize.

There are two main types:

1. Max Priority Queue: The element with the highest priority is dequeued first. It’s commonly used when you need to process the most important or largest element first.
2. Min Priority Queue: The element with the lowest priority is dequeued first. It’s useful for problems like finding the smallest element or processing tasks with the least urgency first.

```python
from queue import PriorityQueue

q = PriorityQueue()

# insert into queue
q.put((2, 'g'))
q.put((3, 'e'))
q.put((4, 'k'))
q.put((5, 's'))
q.put((1, 'e'))
q.put((7, 'z'))

while not q.empty():
    print(q.get())
print(q)

# remove and return 
# lowest priority item
# print(q.get())
# print(q.get())

# check queue size
print('Items in queue :', q.qsize())

# check if queue is empty
print('Is queue empty :', q.empty())

# check if queue is full
print('Is queue full :', q.full())
```
