### Recursion - String

```python
# Remove a character from a string
def remove_char(s, char):
    if len(s) == 0 or s=='':
        return s
    small_answer = remove_char(s[1:], char)
    if s[0] == char:
        return small_answer
    else:
        return s[0] + small_answer
    
remove_char("hello world", "o")
```

```python
# palindrome or not
def palindrome(s):
    if len(s) == 0 or len(s) == 1 or s=='':
        return True
    if s[0] != s[-1]:
        return False
    return palindrome(s[1:-1])
```

```python
def palindrome(s):
    return palindrome_helper(s, 0, len(s)-1)

def palindrome_helper(s, i, j):
    if i>=j:
        return True
    if s[i] != s[j]:
        return False
    return palindrome_helper(s, i+1, j-1)
print(palindrome("racedcar"))
```

### SubString

In Substring, Ordering is maintained and it should be continous.

Ex: For 'abc', substrings are [a, b, c, ab, bc, abc]

### SubSequence

In SubSequence, Order is maintained and they can be non-continous as well

Ex: For 'abc', subsequence are [a, b, c, ab, bc, ac, abc, @] 

```python
def subsequence(s):
    if len(s) == 0:
        return [""]
    curr_char = s[0]
    small_answer = subsequence(s[1:])
    print(f"small_answer: {small_answer}")
    answer = []
    answer.extend(small_answer) 
    for eachPermutation in small_answer:
        answer.append(curr_char+eachPermutation)
    return answer

subsequence("abcd")
        
```

```python
# Print the subsequences of a string
def print_subsequences(s, takenSoFar):
    if len(s) == 0:
        print(takenSoFar)
        return
    currentChar = s[0]
    smallInput = s[1:]
    print_subsequences(smallInput, takenSoFar+currentChar)
    print_subsequences(smallInput, takenSoFar)
    return

print_subsequences("abc", '')
```

### Permutations of the String

```python
# print all the permutations of a string for example abc, the permutations are abc, acb, bac, bca, cab, cba
def print_permutations(s, takenSoFar):
    if len(s) == 0:
        print(takenSoFar)
        return
    our_char = s[0]
    small_input = s[1:]
    for i in range(len(takenSoFar)+1):
        new_takenSoFar = takenSoFar[:i] + our_char + takenSoFar[i:]
        print(f"new_takenSoFar: {new_takenSoFar}")
        print_permutations(small_input, new_takenSoFar)
    return

print_permutations("abc", "")
            
```

```python
# Return all the permutations of a string
def return_permutations(s):
    if len(s) == 0:
        return [""]
    curr_char = s[0]
    small_answer = return_permutations(s[1:])
    answer = []
    for each_permutation in small_answer:
        for i in range(len(each_permutation)+1):
            new_permutation = each_permutation[:i] + curr_char + each_permutation[i:]
            answer.append(new_permutation)
    return answer

return_permutations("abc")
    
```

Mapping of digits to letters (just like old telephone buttons). Note that 1 does not map to any letters

```python
keys = {'1': '', '2': 'abc', '3': 'def', '4': 'ghi', '5': 'jkl', '6': 'mno', '7': 'pqrs', '8': 'tuv', '9': 'wxyz'}

def return_keypad(input):
    if len(input) == 0:
        return [""]
    curr_char = input[0]
    small_output = return_keypad(input[1:])
    answer = []
    keypad_letters = keys[curr_char]
    for each_letter in keypad_letters:
        for each_permutation in small_output:
            answer.append(each_letter+each_permutation)
    return answer

return_keypad("23")
```

## Return all the codes

Every number has a character associated with it.

1 -> a, 2 -> b, 3 -> c, 4 -> d, 5 -> e, 6 -> f 

Input: 1123

Output: aabc, RW, alc, rbc, aaw 



```python

def get_char(digit):
    return chr(ord('a') + digit - 1)


def return_codes(input):
    if len(input) == 0:
        return [""]
    single_digit = input[0:1]
    double_digit = input[0:2]
    mainAnswer = []
    answer_without_first_digit = return_codes(input[1:])

    for each_permutation in answer_without_first_digit:
        mainAnswer.append(get_char(int(single_digit)) + each_permutation)

    if int(double_digit) <= 26:
        answer_without_first_two_digits = return_codes(input[2:])
        for each_permutation in answer_without_first_two_digits:
            mainAnswer.append(get_char(int(double_digit)) + each_permutation)
    return mainAnswer

return_codes("1123")

    
```

```python
## Print all the codes

def print_codes(input, takenSoFar):
    if len(input) == 0:
        print(takenSoFar)
        return
    single_digit = input[0:1]
    double_digit = input[0:2]
    print_codes(input[1:], takenSoFar + get_char(int(single_digit)))
    if int(double_digit) <= 26:
        print_codes(input[2:], takenSoFar + get_char(int(double_digit)))

print_codes("1123", "")
```

### Tower of Hanoi

```python
def tower_of_hanoi(n, source, destination, auxiliary):
    if n == 0:
        return
    if n == 1:
        print(f"Move disk 1 from {source} to {destination}")
        return
    tower_of_hanoi(n-1, source, auxiliary, auxiliary)
    print(f"Move disk {n} from {source} to {destination}")
    tower_of_hanoi(n-1, auxiliary, source, destination)

tower_of_hanoi(4, 'S', 'D', 'A')
```
