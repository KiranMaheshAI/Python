### Searching Using Recursion

#### Linear Search

```python
# Linear search using recursion

def linear_search_using_recursion(l1, x, index):
    # Base case
    if len(l1) == index:
        return False
    
    ansFromRecursion = linear_search_using_recursion(l1, x, index+1)
    return l1[index] == x or ansFromRecursion

linear_search_using_recursion([1, 2,3,4,5,6], 4, 0)
```

```python
# Linear search using recursion - Tail Recursion
def linear_search_using_recursion(l1, x, index):
    # Base case
    if len(l1) == index:
        return False

    if l1[index] == x:
        return True
    
    return linear_search_using_recursion(l1, x, index+1)

linear_search_using_recursion([1, 2,3,4,5,6], 4, 0)
```

#### Binary Search

```python
# Using Helper function
def BinarySearchUsingRecursionHelper(l1, x, s, e):
    if s>e:
        return False
    
    m = int(s + (e-s)//2)

    if l1[m] == x:
        return True
    
    if x > l1[m]:
        return BinarySearchUsingRecursionHelper(l1, x, m+1, e)
    return BinarySearchUsingRecursionHelper(l1, x, s, m-1)

def BinarySearchUsingRecursion(l1, x):
    return BinarySearchUsingRecursionHelper(l1, x, 0, len(l1)-1)

print(BinarySearchUsingRecursion([1,2,3,4,5,6,7,7,7,8,8,8,10], 9))
```
