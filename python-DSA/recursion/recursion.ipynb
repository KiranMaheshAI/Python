### Recursion:

Calling a function by itself.

Recursive function contains two key parts:
1. Base Case: The stopping condition that prevents infinite recursion.
2. Recursive Case: The part of the function where it calls itself with modified parameters.

```python
def factorial(n):
    if n == 0:  # Base case
        return 1
    else:       # Recursive case
        return n * factorial(n - 1)

print(factorial(5))
```

#### Types of Recursion in Python
Recursion can be broadly classified into two types: tail recursion and non-tail recursion. The main difference between them is related to what happens after recursive call.

Tail Recursion: The recursive call is the last thing the function does, so nothing happens after it returns. Some languages can optimize this to work like a loop, saving memory. We should use accumulator, as my result is getting accumulated.

Head Recursion: The function does more work after the recursive call returns, so it can’t be optimized into a loop.

```python
def tail_fact(n, acc=1):
    # Base case
    if n == 0:
        return acc
    # Tail recursive call with an accumulator
    accumulator = acc * n
    return tail_fact(n-1, accumulator)

def head_fact(n):
    # Base case
    if n == 1:
        return 1
    # Non-tail recursive call because the multiplication happens after the call
    small_ans = head_fact(n-1)
    ans = n * small_ans
    return ans

# Example usage
print(tail_fact(5))  
print(head_fact(5))
```

#### When to Avoid Recursion
1. When the problem can be solved easily with loops.
2. When recursion depth is large enough to risk a stack overflow.
3. When performance is critical and function call overhead matters.
