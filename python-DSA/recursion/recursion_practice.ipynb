### Fibonicci Series

Fibonicci series is Sum of last 2 number 

```python
def fibonicci_series(n):
    if n == 0:
        return 1
    if n == 1:
        return 1
    return fibonicci_series(n-1) + fibonicci_series(n-2)

fibonicci_series(5)
```

```python
def fibonacci(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    if n == 2:
        return 1

    last = fibonacci(n-1)
    second_last = fibonacci(n-2)
    ans = last + second_last
    return ans
    
fibonacci(5)
```

### Print count from 1 to n

```python
def count_to_n(n):
    if n==1:
        return [1]
    return count_to_n(n-1)+[n]

count_to_n(5)
```

### Print count from n to 1

```python
def count_to_n(n):
    if n==1:
        return [1]
    return [n]+count_to_n(n-1)

count_to_n(5)
```

### Check if array is sorted

```python
# Head Recursion
def checksorted(l1):
    if len(l1) == 1:
        return True
    ans = checksorted(l1[1:])
    if l1[0] >= l1[1]:
        return False
    return ans
checksorted([1, 2, 3, 5, 6])
```

```python
# Tail Recursion
def checksorted2(l1):
    if len(l1) == 1 or len(l1) == 0:
        return True

    if l1[0] >= l1[1]:
        return False
    return checksorted(l1[1:])

checksorted([1, 2, 3, 5, 6])
```

### Sum of array

```python
def sum_of_array(l1):
    if len(l1) == 1:
        return l1[0]
    return l1[0] + sum_of_array(l1[1:])

sum_of_array([1, 2, 3, 4, 5])
```

```python
# Tail Recursion
def sum_of_array2(l1, accumulator = 0):
    if len(l1) == 1:
        return accumulator + l1[0]
    accumulator += l1[0]
    print(accumulator)
    return sum_of_array2(l1[1:], accumulator)

sum_of_array2([1, 2, 3, 4, 5])
```

### Find index of the element in array

```python
def find_last_index(l1, x):
    if len(l1) == 0:
        return -1
    if l1[-1] == x:
        return len(l1) - 1
    return find_last_index(l1[:-1], x)

print(find_last_index([1, 2, 3, 4, 5], 9))

### Find first index of the element in array
def find_first_index(l1, x):
    if len(l1) == 0:
        return -1
    if l1[0] == x:
        return 0
    return_index = find_first_index(l1[1:], x)
    if return_index == -1:
        return -1
    return 1+return_index

print(find_first_index([1, 2, 3, 4, 5], 4))
```

```python
### Find all indices of the element in array
def find_all_indices(l1, x):
    if len(l1) == 0:
        return []
    ind = find_all_indices(l1[1:], x)
    if l1[0] == x:
        return [0]+[i+1 for i in ind]
    else:
        return [i+1 for i in ind]

find_all_indices([1, 2, 3, 4, 3, 5, 3, 3, 3], 3)
```

```python

### Find all indices of the element in array
def find_all_indices2(l1, x, index = 0):
    if index == len(l1):
        return []
    if l1[index] == x:
        return [index] + find_all_indices2(l1, x, index + 1)
    return find_all_indices2(l1, x, index + 1)

find_all_indices2([1, 2, 3, 4,2, 5], 2)
```

```python
# Getting the list of accurence of element in the list
def find_all_indices3(arr, element):
    def helper(index):
            # Base case: if index is equal to the length of the array, return empty list
            if index == len(arr):
                return []
            
            # Check if the current index contains the element
            if arr[index] == element:
                # Include the current index and proceed with the next index
                return [index] + helper(index + 1)
            else:
                # Proceed with the next index without including the current index
                return helper(index + 1)
    
    # Start the recursive search from index 0
    return helper(0)

find_all_indices3([1, 2, 3, 4, 2, 5], 2)
```
