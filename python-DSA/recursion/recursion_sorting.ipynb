### Sorting Using Recursion

```python
# Merge Sort Using Recursion
def merge(l1, s, m, e):
    i = s
    j = m+1
    ans = []
    while(i<=m and j<=e):
        if l1[i] < l1[j]:
            ans.append(l1[i])
            i+=1
        elif l1[i] > l1[j]:
            ans.append(l1[j])
            j+=1
        elif l1[i] == l1[j]:
            ans.append(l1[i])
            ans.append(l1[j])
            i+=1
            j+=1
        
    while(i<=m):
        ans.append(l1[i])
        i+=1
    while(j<=e):
        ans.append(l1[j])
        j+=1

    startofAns = 0
    startofMyList = s
    while(startofMyList<=e):
        l1[startofMyList] = ans[startofAns]
        startofAns+=1
        startofMyList+=1
    return

def mergeSortHelper(l1, s, e):
    if s>=e:
        return
    m= s+ (e-s)//2

    mergeSortHelper(l1, s, m)
    mergeSortHelper(l1, m+1, e)
    print(f"mergeSortHelper({l1}, {s}, {m}, {e})")
    merge(l1, s, m, e)
    return

l1 = [5, 4, 2, 8, 7, 1, 10, 11, 9, 6, 4]
print(mergeSortHelper(l1, 0, 5))
print(l1)
```

```python
# Merge sort - Other way
def merge_sort(arr):
    # Base case: if the array has 0 or 1 element, it is already sorted
    if len(arr) <= 1:
        return arr

    # Divide the array into two halves
    mid = len(arr) // 2
    left_half = arr[:mid]
    right_half = arr[mid:]

    # Recursively sort the two halves
    left_half = merge_sort(left_half)
    right_half = merge_sort(right_half)

    # Merge the sorted halves
    return merge(left_half, right_half)

def merge(left, right):
    merged_array = []
    i = 0  # Pointer for the left array
    j = 0  # Pointer for the right array

    # Compare elements from both arrays and add the smaller one to the merged array
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            merged_array.append(left[i])
            i += 1
        else:
            merged_array.append(right[j])
            j += 1

    # Add any remaining elements from the left array
    while i < len(left):
        merged_array.append(left[i])
        i += 1

    # Add any remaining elements from the right array
    while j < len(right):
        merged_array.append(right[j])
        j += 1

    return merged_array

# Example usage:
my_array = [12, 11, 13, 5, 6, 7]
sorted_array = merge_sort(my_array)
print(f"Original array: {my_array}")
print(f"Sorted array: {sorted_array}")
```

#### Quick Sort (Pivot)

```python
def partitionFunction(l1, s, e):
    # First step: Get the pivot to the right position
    pivot = l1[e]
    i = s
    rightPosition = s
    while(i<=e-1):
        if l1[i] < pivot:
            rightPosition += 1
        i += 1
    
    l1[rightPosition], l1[e] = l1[e], l1[rightPosition]
    pivot = rightPosition
    # Now make sure that everything smaller than pivot to the left and greater to the right.

    start =s
    end = e

    while(start<pivot and end>pivot):
        if l1[start] < pivot:
            start += 1
        elif l1[end] >= pivot:
            end-=1
        else:
            l1[start], l1[end] = l1[end], l1[start]
            start+=1
            end-=1

    return pivot


# l1 = [3, 7, 8, 3, 2, 5, 4]
# partitionFunction(l1, 0, len(l1)-1)
# print(l1)

def quick_sort(l1, s, e):
    if s>=e:
        return
    
    pivotIndex = partitionFunction(l1, s, e)
    quick_sort(l1, s, pivotIndex-1)
    quick_sort(l1, pivotIndex+1, e)
    return


l1 = [7, 8, 3, 2, 5, 4]
quick_sort(l1, 0, (len(l1)-1))
print(l1)
```
