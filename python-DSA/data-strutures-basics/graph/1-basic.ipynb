### Graphs

A graph is a collection of nodes and edges.

Nodes are the vertices of the graph.
Edges are the connections between the nodes.

Real life examples:
1. Social Networks
2. Maps
3. Transportation
4. Internet
5. Computer Networks

Industry Use cases:
1. Google Maps
2. Facebook
3. Instagram
4. Twitter
5. LinkedIn
6. Search Engines
7. Recommendation Systems
8. Network Security
9. Routing Algorithms
10. Scheduling

Tree is a special case of graph where:
1. Every node is connected to each other in someway.
2. There is no cycle in the graph.
3. There is only one path between any two nodes.

### Terminologies in Graph

1. Node (Vertex) : Individual element in a graph.
2. Edge: Connection between Two vertices, it can be Unidirectional or Bidirectional.
3. Adjacent Vertex: Two vertices are said to be adjacent if they are connected by an edge.
4. Degree: Number of edge connected to a node/vertex. It can have InDegree and OutDegree.
5. Path: Path is a sequence of nodes from source node to destination node, it can be Cyclic or Acyclic path.
6. Cycle: If path starts and end at the same vertex. It is a closed path.

Different Types of Graphs:

1. Undirected Graph: Graph where edge have no directions. Ex: Friends on facebook.
2. Directed Graph: Graph where edges has a direction. Ex: Follow on Instagram, Goodle maps.
3. Weighted Graph: graph where edges have weights which are non integers.
4. Unweighted Graph: Graph where edges have no weight.
5. Cyclic Graph: Graph which contains atleast one cycle.
6. Acyclic Graph: Graph which does not contains any cycle.
7. Connected Graph(Dense Graph): Graph where each node is connected by every other node by atleast one edge.
8. Labelled Graph: Graph where we are adding label to any node or edge.
9. Connected Graph: Graph where each node is connected where one node can go to every other node.  

```python
class GraphUsingEdgeList:
    def __init__(self):
        self.vertices = []
        self.edge_list = []

    def add_vertex(self, vertex):
        if vertex not in self.vertices:
            self.vertices.append(vertex)
        else: 
            print("Vertex already exists")

    def add_edge(self, source, destination, weight=1):
        if source in self.vertices and destination in self.vertices:
            self.edge_list.append((source, destination, weight))
        else:
            print("Vertex not found")

    def print_graph(self):
        for edge in self.edge_list:
            print(edge)

    def display_graph(self):
        for vertex in self.vertices:
            print(vertex, end=" -> ")
            for edge in self.edge_list:
                if edge[0] == vertex:
                    print(edge[1], end=" ")
            print()

graph = GraphUsingEdgeList()
graph.add_vertex("A")
graph.add_vertex("B")
graph.add_vertex("C")
graph.add_vertex("D")
graph.add_edge("A", "B")
graph.add_edge("B", "C")
graph.print_graph()
graph.display_graph()
```

#### Adjacency List

```python
class GraphUsingAdjacencyList:
    def __init__(self):
        self.vertices = []
        self.adjacency_list = {}

    def add_vertex(self, vertex):
        if vertex not in self.vertices:
            self.vertices.append(vertex)
            self.adjacency_list[vertex] = []
        else:
            print("Vertex already exists")

    def add_edge(self, source, destination, weight=1):
        if source in self.vertices and destination in self.vertices:
            self.adjacency_list[source].append((destination, weight))
            self.adjacency_list[destination].append((source, weight))
        else:
            print("Vertex not found")

    def display_graph(self):
        for vertex in self.vertices:
            print(vertex, end=" -> ")
            for edge in self.adjacency_list[vertex]:
                print(edge, end=" ")
            print()

graph = GraphUsingAdjacencyList()
graph.add_vertex("A")
graph.add_vertex("B")
graph.add_vertex("C")
graph.add_vertex("D")
graph.add_edge("A", "B")
graph.add_edge("B", "C")
graph.display_graph()
```

### Adjacency Matrix

Adjacency Matrix is a square matrix used to represent a finite graph. The elements of the matrix indicate whether pairs of vertices are adjacent or not in the graph. An adjacency matrix is a simple and straightforward way to represent graphs and is particularly useful for dense graphs. For a graph with V vertices, the adjacency matrix A is an V X V matrix or 2D array.

Use Cases: 
1. Data Structures: Efficiently represents graph connections in computer science.
2. Algorithm Analysis: Simplifies graph traversal (like Breadth-First Search, Depth-First Search) and analysis of graph properties.
3. Weighted Graphs: Stores edge weights instead of just 1s. 

Disadvantages of Adjacency Matrix:
1. Space inefficient for Sparse Graphs: Takes up O(V* V) space even if the graph is sparse.
2. Costly Insertions and Deletions: Adding or deleting a vertex can be costly as it requires resizing the matrix.
3. Slow Traversals: Graph traversals like DFS, BFS takes O(V * V) time to visit all the vertices whereas Adjacency List takes only O(V + E).

```python
class GraphAdjacencyMatrix:
    def __init__(self, num_vertices):
        self.num_vertices = num_vertices
        self.vertices = [None] * num_vertices
        self.adjacency_matrix = [[0] * num_vertices for _ in range(num_vertices)]

    def add_vertex(self, index, label):
        if 0 <= index < self.num_vertices:
            self.vertices[index] = label
        else:
            print("Index OOB")

    def add_edge(self, source, destination, weight=1):
        if 0 <= source < self.num_vertices and 0 <= destination < self.num_vertices:
            self.adjacency_matrix[source][destination] = weight
            self.adjacency_matrix[destination][source] = weight # for undirected graph
        else:
            print("Index OOB")

    def display_graph(self):
        for index, label in enumerate(self.vertices):
            if label is not None:
                print(f"Vertex index : {index}, label : {label}")
        for row in self.adjacency_matrix:
            print(row)

graph = GraphAdjacencyMatrix(4)
graph.add_vertex(0, "A")
graph.add_vertex(1, "B")
graph.add_vertex(2, "C")
graph.add_vertex(3, "D")
graph.add_edge(0, 1)
graph.add_edge(1, 2)
graph.add_edge(2, 3)
graph.display_graph()
```
