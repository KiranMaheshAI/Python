## LinkedList

Linkedlist is a linear data structiure like a array or list. Linked list add elements as per our requirements.

Linkedlist randomly allocated space in memory.

LinkedList stores data and address of next element.

```python
# Create a Node LinkedList

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

first = Node(10)
second = Node(20)
third = Node(30)

first.next = second
second.next = third

head = first
print(head.data)
print(head.next.data)
print(head.next.next.data)
```

```python
# Print the linked list
current = head
while current:
    print(current.data)
    current = current.next
```

```python
# Take Inputs to the linked list
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

def printLL(head):
    while head is not None:
        print(head.data, end=" ")
        head = head.next
    print()

head = None
temp = None
def takeInput():
    global head
    global temp
    value = int(input("Enter the value of the node: "))
    while value != -1:
        if head is None:
            head = Node(value)
            temp = head
        else:
            temp.next = Node(value)
            temp = temp.next
        value = int(input("Enter the value of the node: "))
    return head
printLL(takeInput())
```

```python
# Length of Linked List
def length(head):
    count = 0
    while head is not None:
        count += 1
        head = head.next
    return count
```

```python
def lengthOfLLRecursive(head):
    if head is None:
        return 0
    return 1 + lengthOfLLRecursive(head.next)
print(lengthOfLLRecursive(takeInput()))
    
```

```python
# Insert At Head of the Linked List
def insertAtHead(head, data):
    newNode = Node(data)
    newNode.next = head
    return newNode

head = insertAtHead(takeInput(), 10)
printLL(head)
        
```

```python
# Insert at tail of the linked list
def insertAtTail(head, data):
    newNode = Node(data)
    if head is None:
        return newNode
    current = head
    while current.next is not None:
        current = current.next
    current.next = newNode
    return head

head = insertAtTail(takeInput(), 10)
printLL(head)
```

```python
# Insert at Tail position Recursively
def insertAtTailRecursively(head, data):
    if head is None:
        return Node(data)
    currentNode = insertAtTailRecursively(head.next, data)
    head.next = currentNode
    return head


head = None
head = insertAtTailRecursively(head, 10000)
head = insertAtTailRecursively(head, 20000)
head = insertAtTailRecursively(head, 30000)
head = insertAtTailRecursively(head, 40000)
head = insertAtTailRecursively(head, 50000)
head = insertAtTailRecursively(head, 60000)
printLL(head)
```

```python
# Insert at i-th position recursively
def insertAtIthPosition(head, i, data):
    if i < 0:
        return head
    if i == 0:
        nextValue = head.next
        head = Node(data)
        head.next = nextValue
        return head
    if head is None:
        return Node(data)
    smallHead = insertAtIthPosition(head.next, i-1, data)
    head.next = smallHead
    return head
        
```

```python
# Delete a node by value
def delete_node_by_value(head, value):
    if head is None:
        return None
    if head.data == value:
        return head.next
    temp = head
    while temp.next is not None and temp.next.data != value:
        temp = temp.next

    if temp.next is not None:
        temp.next = temp.next.next
    return head
```

## Search By Node

```python
def searchByNode(head, value):
    current = head
    while current is not None:
        if current.data == value:
            return True
        current = current.next
    return False
```

```python
def searchByNodeRecursively(head, value):
    if head is None:
        return False
    if head.data == value:
        return True
    return searchByNodeRecursively(head.next, value)

```

```python
# search by node iteratively
def searchByNodeIteratively(head, value):
    current = head
    while current is not None:
        if current.data == value:
            return True
        current = current.next

print(searchByNodeIteratively(head, 10))
```

```python
# search a node by value recursively and return the index
def searchByNodeRecursively(head, value):
    if head is None:
        return -1
    if head.data == value:
        return 0
    index = searchByNodeRecursively(head.next, value)
    return index + 1

print(searchByNodeRecursively(head, 10))
# search a node by value iteratively and return the index
def searchByNodeIteratively(head, value):
    current = head
    index = 0
    while current is not None:
        if current.data == value:
            return index
        current = current.next
        index += 1
    return -1

print(searchByNodeIteratively(head, 10))
```

### Reverse a LL

```python
def reverse(head):
    current = head
    previous = None
    while current is not None:
        next = current.next
        current.next = previous
        previous = current
        current = next
    head = previous

def reverseRecursively(head):   
    if head is None:
        return
    reverseRecursively(head.next)
    head.next.next = head
    head.next = None
    head = head.next
    return head

def reverseIteratively(head):
    current = head
    previous = None
    while current is not None:
        next = current.next
        current.next = previous
        previous = current
        current = next
    head = previous
    return head

```
