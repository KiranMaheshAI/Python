### 1. Minimum steps to reduce n to 1

Minimum number of steps required to reduce the number to 1. Below are the steps (i) Subtract 1  (ii) Divide by 2 (iii) Divide by 3 

```python
# Recursion approach
def minimum_step_required(n):
    if n == 1:
        return 0
    if n % 2 == 0:
        return 1 + minimum_step_required(n//2)
    elif n % 3 == 0:
        return 1 + minimum_step_required(n//3)
    else:
        return 1 + minimum_step_required(n-1)

print(minimum_step_required(10))

# Memorization
def minimum_step_required_memo(n, memo):
    if n == 1:
        return 0
    if memo[n] is not None:
        return memo[n]
    if n % 2 == 0:
        memo[n] = 1 + minimum_step_required_memo(n//2, memo)
    elif n % 3 == 0:
        memo[n] = 1 + minimum_step_required_memo(n//3, memo)
    else:
        memo[n] = 1 + minimum_step_required_memo(n-1, memo)
    return memo[n]

print(minimum_step_required_memo(10, [None] * 11))

# DP Approach
def minimum_step_required_dp(n):
    dp = [0] * (n+1)
    dp[1] = 0
    for i in range(2, n+1):
        if i % 2 == 0:
            dp[i] = 1 + dp[i//2]
        elif i % 3 == 0:
            dp[i] = 1 + dp[i//3]
        else:
            dp[i] = 1 + dp[i-1]
    return dp[n]

print(minimum_step_required_dp(10))
```

### 2. Count Balanced BTS

Number of balanced binary trees with the given height. We can calculate the count of balanced binary trees with the formula:

For the height of 2, we calculate using the formula: (height - 1) * (height - 1) and a product of 2 * (height-2) * (height - 1)


```python
# Recursion approach
def count_balanced_bts(height):
    if height == 0 or height == 1:
        return 1
    h1 = count_balanced_bts(height-1)
    h2 = count_balanced_bts(height-2)
    ans = h1 * h1 + 2 * h1 * h2
    return ans

print(count_balanced_bts(2))

# Memorization
def count_balanced_bts_memo(height, memo):
    if height == 0 or height == 1:
        return 1
    if memo[height] is not None:
        return memo[height]
    h1 = count_balanced_bts_memo(height-1, memo)
    h2 = count_balanced_bts_memo(height-2, memo)
    ans = h1 * h1 + 2 * h1 * h2
    memo[h1] = h1
    memo[h2] = h2
    memo[height] = ans
    return ans

print(count_balanced_bts_memo(2, [None] * 3))

# DP Approach
def count_balanced_bts_dp(height):
    dp = [0] * (height+1)
    dp[0] = 1
    dp[1] = 1
    for i in range(2, height+1):
        dp[i] = (dp[i-1] * dp[i-1]) + (2 * dp[i-1] * dp[i-2])
    return dp[height]

print(count_balanced_bts_dp(2))
```

### 3. Minimum number of squares

Minimum number of squares to cover the given number.

```python
import math
# Recursion approach
def minimum_squares(n):
    if n == 0:
        return 0
    min_squares = float('inf')
    for i in range(1, int(math.sqrt(n)) + 1):
        square = i * i
        if square > n:
            break
        print(n, square)
        min_squares = min(min_squares, 1 + minimum_squares(n - square))
    return min_squares

print(minimum_squares(12))
```

```python
# Memorixation approach
def minimum_squares_memo(n, memo):
    if n == 0:
        return 0
    if memo[n] is not None:
        return memo[n]
    min_squares = float('inf')
    for i in range(1, int(math.sqrt(n)) + 1):
        square = i * i
        if square > n:
            break
        min_squares = min(min_squares, 1 + minimum_squares_memo(n - square, memo))
    memo[n] = min_squares
    print(memo)
    return min_squares

print(minimum_squares_memo(12, [None] * 13))

# DP approach
def minimum_squares_dp(n):
    dp = [0] * (n+1)
    dp[0] = 0
    for i in range(1, n+1):
        min_squares = float('inf')
        for j in range(1, int(math.sqrt(i)) + 1):
            square = j * j
            if square > i:
                break
            min_squares = min(min_squares, 1 + dp[i - square])
        dp[i] = min_squares
    return dp[n]

print(minimum_squares_dp(12))
```

## 4. Minimum cost path 

2D array : [[1, 2, 3][4, 8, 2][1, 5, 3]]

Directions in which we can move 
1) Right (0, 0) => (0, 1)
2) Down (0, 0) => (1, 0)
3) Diagnolly (0, 0) => (1, 1)

```python
# Recursion approach
def minimum_cost_path(arr):
    return minimum_cost_path_helper(arr, 2, 2)

def minimum_cost_path_helper(arr, i, j):
    if i == 0 and j == 0:
        return arr[i][j]
    if i < 0 or j < 0:
        return float('inf')
    return arr[i][j] + min(minimum_cost_path_helper(arr, i-1, j), minimum_cost_path_helper(arr, i, j-1), minimum_cost_path_helper(arr, i-1, j-1))

print(minimum_cost_path([[1, 2, 3], [4, 8, 2], [1, 5, 3]]))


# Memorization approach
def minimum_cost_path_memo(arr, memo):
    return minimum_cost_path_helper_memo(arr, 2, 2, memo)

def minimum_cost_path_helper_memo(arr, i, j, memo):
    if i == 0 and j == 0:
        return arr[i][j]
    if memo[i][j] is not None:
        return memo[i][j]
    if i < 0 or j < 0:
        return float('inf')
    memo[i][j] = arr[i][j] + min(minimum_cost_path_helper_memo(arr, i-1, j, memo), minimum_cost_path_helper_memo(arr, i, j-1, memo), minimum_cost_path_helper_memo(arr, i-1, j-1, memo))
    return memo[i][j]

print(minimum_cost_path_memo([[1, 2, 3], [4, 8, 2], [1, 5, 3]], [[None] * 3 for _ in range(3)]))

# DP approach
def minimum_cost_path_dp(arr):
    dp = [[0] * len(arr[0]) for _ in range(len(arr))]
    dp[0][0] = arr[0][0]
    for i in range(1, len(arr)):
        dp[i][0] = dp[i-1][0] + arr[i][0]
    for j in range(1, len(arr[0])):
        dp[0][j] = dp[0][j-1] + arr[0][j]
    for i in range(1, len(arr)):
        for j in range(1, len(arr[0])):
            dp[i][j] = arr[i][j] + min(dp[i-1][j], dp[i][j-1], dp[i-1][j-1])
    return dp[len(arr)-1][len(arr[0])-1]

print(minimum_cost_path_dp([[1, 2, 3], [4, 8, 2], [1, 5, 3]]))
```

```python
# DP approach through row wise
def minimum_cost_path_row_wise(arr):
    dp = [[0] * len(arr[0]) for _ in range(len(arr))]
    dp[0][0] = arr[0][0]
    for row in range(len(arr)):    
        for col in range(0, len(arr[0])):
            if row == 0 and col == 0:
                continue
            cost_left = dp[row][col-1] if col > 0 else float("inf")
            cost_top = dp[row-1][col] if row > 0 else float("inf")
            cost_diag = dp[row-1][col-1] if row >0 and col >0 else float("inf")
            ans = arr[row][col] + min(cost_left, cost_top, cost_diag)
            dp[row][col] = ans
    return dp[len(arr)-1][len(arr[0])-1]

print(minimum_cost_path_row_wise([[1, 2, 3], [4, 8, 2], [1, 5, 3]]))


# DP approach through col wise
def minimum_cost_path_row_wise(arr):
    dp = [[0] * len(arr[0]) for _ in range(len(arr))]
    dp[0][0] = arr[0][0]
    for col in range(len(arr)):    
        for row in range(0, len(arr[0])):
            if row == 0 and col == 0:
                continue
            cost_left = dp[row][col-1] if col > 0 else float("inf")
            cost_top = dp[row-1][col] if row > 0 else float("inf")
            cost_diag = dp[row-1][col-1] if row >0 and col >0 else float("inf")
            ans = arr[row][col] + min(cost_left, cost_top, cost_diag)
            dp[row][col] = ans
    return dp[len(arr)-1][len(arr[0])-1]

print(minimum_cost_path_row_wise([[1, 2, 3], [4, 8, 2], [1, 5, 3]]))
```

### 5. Longest common subsequence

```python
def longest_common_subsequence(str1, str2):
    return longest_common_subsequence_helper(str1, str2, 0, 0)

def longest_common_subsequence_helper(str1, str2, i, j):
    if i == len(str1) or j == len(str2):
        return 0
    if str1[i] == str2[j]:
        return 1 + longest_common_subsequence_helper(str1, str2, i+1, j+1)
    else:
        return max(longest_common_subsequence_helper(str1, str2, i+1, j), longest_common_subsequence_helper(str1, str2, i, j+1))

print(longest_common_subsequence("ABCDGH", "AEDFHR"))
```

## 6. length of the longest strictly increasing subsequence

You need the length of the longest strictly increasing subsequence (not necessarily contiguous).

Example:
[10, 9, 2, 5, 3, 7, 101, 18]

One LIS → [2, 3, 7, 101] → length = 4

```python
def longest_increasing_subsequence(arr):
    if not arr:
        return 0
    dp = [1]* len(arr)
    for i in range(len(arr)):
        for j in range(i):
            if arr[j] < arr[i]:
                dp[i] = max(dp[i], dp[j]+1)
    return max(dp)

print(longest_increasing_subsequence([10, 9, 2, 5, 3, 7, 101, 18]))
```

### Alternative way
Using bisect module in python

The bisect module in Python's standard library provides functions for manipulating sorted lists using the efficient binary search algorithm. It is primarily used to find the correct insertion point for a new element or to insert the element while maintaining the sort order, operating in O(log n) time for the search part. 

```python
import bisect

def lengthOfLIS(nums):
    tails = []

    for x in nums:
        pos = bisect.bisect_left(tails, x)
        if pos == len(tails):
            tails.append(x)
        elif pos == 0:
            print(f"{x} in {pos} in the {tails}")
            tails.clear()
            tails.append(x)
        else:
            print(f"{x} in {pos} in the {tails}")
            while(pos<=len(tails)-1):
                tails.pop()
            tails.append(x)
    return len(tails)

print(lengthOfLIS([0,1,0,3,2,3]))
```

### Nth Tribonacci number
The Tribonacci sequence is defined as follows:

T0 = 0

T1 = 1

T2 = 1

For n >= 0, Tn+3 = Tn + Tn+1 + Tn+2



```python
def tribonacci_number(n):
    if n == 0:
        return 0
    if n == 1 or n == 2:
        return 1
    dp = [0, 1, 1]
    for i in range(3, n+1):
        val = dp[i-1]+ dp[i-2]+ dp[i-3]
        dp.append(val)
    print(dp)
    return dp[-1]

tribonacci_number(4)
```
