## Dynamic Programming(DP)

Dynamic programming is an optimization technique used to solve problems by:

-> Breaking them into smaller subproblems
-> Solving each sub problem once
-> storing their solutions

Whenever we solve any recursive problem, we are making recursive calls to some values many times(Recursion). So, the time complexity is 2(power n).

To reduce the repetative calls we add Memorization to the values which will reduce the number of calls to O(n). This approach is called Top Down Approach. Here we go with (recursion + Memorization).

There is another way for calculating the answer from bottom, where we don't use recursion. That aproach is called Bottom Up Approach or Tabulation or Dynamic programming.

So, any recursive problem is solved in three ways:
1. Recursion
2. Recursion + Memorization
3. DP Approach

###  Fibonacci series

```python
# Recursive Approach
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

print(fibonacci(5))
```

```python
# Memorization
def fibonacci(n, memo):
    if n <= 1:
        return n
    if memo[n] is not None:
        return memo[n]
    memo[n] = fibonacci(n-1, memo) + fibonacci(n-2, memo)
    return memo[n]

print(fibonacci(5, [None] * 6))
```

```python
# DP Approach
def fibonacci(n):
    dp = [0] * (n+1)
    dp[0] = 0
    dp[1] = 1
    for i in range(2, n+1):
        dp[i] = dp[i-1] + dp[i-2]
    return dp[n]

print(fibonacci(5))
```
