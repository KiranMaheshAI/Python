### Sliding Window Algorithm

The sliding window algorithm is a powerful algorithmic technique used to efficiently solve problems involving contiguous subarrays or substrings within a larger sequence (like an array or a string).

The key idea is to reuse the results of the previous window's computation: 
1. When the window moves, the element leaving the window from the left is removed from the current calculation.
2. The new element entering the window from the right is added to the calculation. 

##### When to Use It
The sliding window technique is applicable to problems that generally have the following characteristics: 
1. The problem involves an ordered data structure like an array or a string.
2. You are looking for an optimal value (maximum, minimum, longest, shortest) within a contiguous subrange.
3. A brute-force solution with nested loops would have a high time complexity, usually O(n²) or O(n³). 

##### Types of Sliding Windows
There are two main variations depending on the problem constraints: 
1. Fixed-Size Window: The size of the window (k) is predefined and remains constant throughout the process. The window slides one element at a time.
2. Variable-Size Window: The window size expands or shrinks dynamically based on certain conditions defined by the problem, such as finding the "longest substring without repeating characters". 

The art of sliding window technical is keep previous calculation to avoid recalculate again, and this will help reduce complexity of code from O(n^2) to O(n).

#### How to Identify Sliding Window Problems?

1. These problems generally require Finding Maximum/Minimum Subarray, Substrings which satisfy some specific condition.
2. The size of the subarray or substring ‘k’ will be given in some of the problems.
3. These problems can easily be solved in O(n2) time complexity using nested loops, using sliding window we can solve these in O(n) Time Complexity.
4. Required Time Complexity: O(n) or O(n log n)


#### Ex: calculate the maximum sum of a subarray having size exactly k.

Input  : arr[] = [5, 2, -1, 0, 3], k = 3
Output : 6
Explanation : We get maximum sum by considering the subarray [5, 2 , -1]

```python
import sys

def maxSum(arr, n, k):
    # Initialize result
    max_sum = -sys.maxsize

    # Consider all blocks starting with i.
    for i in range(n - k + 1):
        current_sum = 0
        for j in range(k):
            current_sum += arr[i + j]

        # Update result if required.
        max_sum = max(current_sum, max_sum)

    return max_sum


if __name__ == "__main__":
    arr = [5, 2, -1, 0, 3]
    k = 3
    n = len(arr)
    print(maxSum(arr, n, k))
```

Second Way

```python
def maxSum(arr, k):
    
    # length of the array
    n = len(arr)

    # n must be greater than k
    if n <= k:
        print("Invalid")
        return -1

    # Compute sum of first window of size k
    window_sum = sum(arr[:k])

    # first sum available
    max_sum = window_sum

    # Compute the sums of remaining windows by
    # removing first element of previous
    # window and adding last element of
    # the current window.
    for i in range(n - k):
        window_sum = window_sum - arr[i] + arr[i + k]
        max_sum = max(window_sum, max_sum)

    return max_sum


if __name__ == "__main__":
    arr = [5, 2, -1, 0, 3]
    k = 3
    print(maxSum(arr, k))
```
