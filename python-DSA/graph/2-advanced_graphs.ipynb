# Traversals in Graphs

Graph traversal is systematically visiting every vertex (node) in a graph, following its edges, to explore connections, find paths, or analyze network structure

There are two types of Traversals in Graph:
1. Depth-First Search(DFS)
2. Breadth First Search (BFS)

Graph traversal is fundamental for many applications, including: 
1. Finding routes (like GPS).
2. Social network analysis (finding friends of friends).
3. Recommendation systems.
4. Detecting cycles in code or networks. 

## Depth First Search

Depth First Search (DFS) is a graph traversal method that starts from a source vertex and explores each path completely before backtracking and exploring other paths. To avoid revisiting nodes in graphs with cycles, a visited array is used to track visited vertices.

Time complexity: O(V + E). We visit every vertex at most once and every edge is traversed at most once (in directed) and twice in undirected.

### Depth first Search - Adjacency Matrix

```python
class GraphAdjacencyMatrix:
    def __init__(self, num_vertices):
        self.num_vertices = num_vertices
        self.vertices = [None] * num_vertices
        self.adjacency_matrix = [[0] * num_vertices for _ in range(num_vertices)]
        
    def add_vertex(self, index, label):
        if 0 <= index < self.num_vertices:
            self.vertices[index] = label
        else:
            print("Index OOB")
            
    def add_edge(self, source, destination, weight=1):
        if 0 <= source < self.num_vertices and 0 <= destination < self.num_vertices:
            self.adjacency_matrix[source][destination] = weight
            self.adjacency_matrix[destination][source] = weight
        else:
            print("Index OOB")
    
    def display_graph(self):
        for index, label in enumerate(self.vertices):
            if label is not None:
                print(f"Vertex index : {index}, label : {label}")
            else:
                print(f"Vertex index : {index}, label : None")
        for row in self.adjacency_matrix:
            print(row)

    def dfs(self, start_vertex):
        visited = [False] * self.num_vertices
        print("Starting DFS from vertex:", self.vertices[start_vertex])
        self._dfs_recursive(start_vertex, visited)

    def _dfs_recursive(self, current_vertex, visited):
        visited[current_vertex] = True
        print(f"Visiting vertex: {self.vertices[current_vertex]}")
        for neighbor in range(self.num_vertices):
            if self.adjacency_matrix[current_vertex][neighbor] != 0 and not visited[neighbor]:
                self._dfs_recursive(neighbor, visited)

graph = GraphAdjacencyMatrix(4)
graph.add_vertex(0, "A")
graph.add_vertex(1, "B")
graph.add_vertex(2, "C")
graph.add_vertex(3, "D")
graph.add_vertex(4, "E")
graph.add_edge(0, 1)
graph.add_edge(0, 2)
graph.add_edge(1, 2)
graph.add_edge(2, 3)
graph.add_edge(1, 4)
graph.display_graph()

graph.dfs(0)
```
