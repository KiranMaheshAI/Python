## Monotonic Stack

A Monotonic Stack is a stack that keeps its elements in a specific order — either always increasing or always decreasing.

In a Monotonic Increasing Stack, each new element you push is larger than the ones already in the stack.

Ex: if you have the numbers 2, 5, 8 in the stack, the next number you add must be greater than 8.

In a Monotonic Decreasing Stack, each new element you push is smaller than the ones in the stack.

Ex: if you have 9, 6, 4 in the stack, the next number must be smaller than 4.

### Pattern 1 : Next Greater Element.

Input: [2, 1, 5, 6, 2, 3]
Output: [5, 5, 6,-1, 3, -1]

```python
# Brute Force

def nextGreaterElementBruteForce(nums: list[int]) -> list[int]:
    n = len(nums)
    result = [0] * n
    
    # For each element
    for i in range(n):
        result[i] = -1  # Default: no greater element found
        
        # Check all elements to the right
        for j in range(i + 1, n):
            if nums[j] > nums[i]:
                result[i] = nums[j]
                break  # Found the next greater, stop searching
    
    return result
nextGreaterElementBruteForce([2, 1, 5, 6, 2, 3])
```

```python
# Best way
def nextGreaterElement(input):
    n =  len(input)
    stack = []
    result = [-1] * n
    for i in range(len(input)):
        if stack and input[i] > input[stack[-1]]:
            while stack:
                value = stack.pop()
                if input[value] < input[i]:
                    result[value] = input[i]
        stack.append(i)
    return result
nextGreaterElement([2, 1, 5, 6, 2, 3])
```
