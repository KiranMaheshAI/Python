## Two Pointers
The two pointers technique is a common algorithmic pattern that uses two references to traverse data structures, typically arrays or strings. 

### Core Concept
Instead of using nested loops (O(n²)), you use two pointers that move through the data structure based on certain conditions, often achieving O(n) time complexity.

#### Use it when:
1. Array is sorted
2. You need pairs / triplets
3. You are dealing with subarrays
4. You need to remove duplicates
5. You need to compare from both ends

##### Common Patterns
1. Opposite Direction (Convergence) : 
Two pointers start at opposite ends and move toward each other.
Use cases: Checking palindromes, pair sum in sorted arrays, container with most water

```python
def two_sum_sorted(arr, target):
    left, right = 0, len(arr) - 1
    
    while left < right:
        current_sum = arr[left] + arr[right]
        if current_sum == target:
            return [left, right]
        elif current_sum < target:
            left += 1
        else:
            right -= 1
    return None

print(two_sum_sorted([2, 3, 5, 6, 7, 8, 9], 12))
```

2. Same Direction (Fast & Slow) : Both pointers start at the same position but move at different speeds.

Use cases: Removing duplicates, partitioning arrays, cycle detection in linked lists

```python
def remove_duplicates(arr):
    if not arr:
        return 0
    
    slow = 0
    for fast in range(1, len(arr)):
        if arr[fast] != arr[slow]:
            slow += 1
            arr[slow] = arr[fast]
    return slow + 1
print(remove_duplicates([1, 2, 3, 3, 4, 5]))
```

## 3Sum – Core Pattern Master
Given an integer array nums, return all unique triplets [nums[i], nums[j], nums[k]] such that:

i ≠ j ≠ k

nums[i] + nums[j] + nums[k] = 0

```python
def threeSum(nums):
    nums.sort()
    result = []
    n = len(nums)
    for i in range(n - 2):
        if i > 0 and nums[i] == nums[i-1]:
            continue
        
        left = i + 1
        right = n - 1
        while left < right:
            total = nums[i] + nums[left] + nums[right]
            if total == 0:
                result.append([nums[i], nums[left] , nums[right]])
                left += 1
                right -= 1

                # Skip Duplicates 
                while left < right and nums[left] == nums[left - 1]:
                    left += 1
                while left < right and nums[right] == nums[right + 1]:
                    right -= 1
            elif total > 0:
                right -= 1
            else:
                left += 1
    return result
threeSum([-1, 0, 1, 2, -1, -4])
```

### Trapping Rain Water – Two Pointer Master Explanation

Given an array height[] where each element represents elevation, compute how much water can be trapped after rain.


```python
def trap(height):
    left = 0
    right = len(height) - 1
    leftMax = 0
    rightMax = 0
    water = 0
    
    while left < right:
        print(f"Left Max is {leftMax}, Right max is {rightMax}, Left is {left} - {height[left]}, Right is {right} - {height[right]}")
        if height[left] < height[right]:
            if height[left] >= leftMax:
                leftMax = height[left]
            else:
                water += leftMax - height[left]
            left += 1
        else:
            if height[right] >= rightMax:
                rightMax = height[right]
            else:
                water += rightMax - height[right]
            right -= 1
        print(f"Water is {water} and left is {left} and right is {right}")
    return water

trap([4,2,0,3,2,5])
```

## Container With Most Water

Given an array height[], each element represents a vertical line.
Find two lines that together with the x-axis form a container that holds the maximum water.


```python
def maxArea(height):
    left  = 0
    right = len(height) - 1
    maxArea = 0
    while left < right:
        area = min(height[left], height[right]) * (right - left)
        if area > maxArea:
            maxArea = area
        if height[left] > height[right]:
            right -= 1
        else:
            left += 1
    return maxArea
print(maxArea([1,8,6,2,5,4,8,3,7]))
```
