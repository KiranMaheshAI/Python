## Sets and Maps (Hash Tables)

Set — stores unique elements, checks membership in O(1). 
Map (Dictionary) — stores key-value pairs, lookups in O(1)

Both are built on hash tables — data structures that use a hash function to map keys to array indices for constant-time access.

### Pattern 1: Removing Duplicates

```python
def remove_duplicates(arr):
    result = []
    for num in arr:
        if num not in result:
            result.append(num)
    return result
print(remove_duplicates([1, 2, 3, 1, 2]))

# Check if All Unique
def has_duplicates(arr):
    return len(arr) != len(remove_duplicates(arr))

has_duplicates([1, 2, 3])
```

## Pattern 2: Two Sum and Variants

Two Sum — find indices of two numbers that sum to target.

```python
def two_sum(arr, target):
    seen = {}

    for i, num in enumerate(arr):
        complement = target - num
        if complement in seen:
            return [seen[complement], i]
        seen[num] = i
    return []

print(two_sum([2, 7, 11, 15], 17))
```

#### Two Sum — Count Pairs:


```python
def count_pairs(arr, target):
    seen = {}
    result = []
    for i, num in enumerate(arr):
        complement = target - num
        if complement in seen:
            result.append([seen[complement], i])
        seen[num] = i
    return result

print(count_pairs([1, 2, 3, 4, 3], 6))


# Another way when duplicates are available
def count_pairs(arr, target):
    seen = {}
    count = 0
    for num in arr:
        complement = target - num
        count += seen.get(complement, 0)
        seen[num] =  seen.get(complement, 0) + 1
    return count

print(count_pairs([1, 2, 3, 4, 3, 2], 6))

# Two Sum - All Unique pairs
def unique_pairs(arr, target):
    seen = set()
    pairs = set()
    for num in arr:
        complement = target - num
        if complement in seen:
            pairs.add(tuple(sorted([num, complement])))
        seen.add(num)
    return [list(p) for p in pairs]
print(unique_pairs([1, 2, 3, 4, 3, 2], 5))
```

### Pattern 3: Frequency Counting

```python
def frequency_character(s):
    freq = {}
    for char in s:
        freq[char] = freq.get(char, 0) + 1
    return freq

# Using Counter
from collections import Counter
def character_frequency(s):
    return Counter(s)

print(character_frequency("hello"))
print(frequency_character("hello"))
```

```python
# First Unique character
def first_unique_char(s):
    freq = character_frequency(s)
    for i, char in enumerate(freq):
        if freq[char] == 1:
            return char
    return -1
print(first_unique_char('hello'))
```

### Top K Frequent

```python
import heapq

# Using Counters
def top_k_frequenct(nums, k):
    freq = Counter(nums)
    return [num for num, count in freq.most_common(k)]

print(top_k_frequenct([1, 1, 1, 2, 2, 3, 3, 4, 4], 3))
```

```python
# Using Hashmap and lists
def top_k_freq(nums, k):
    freq = {}
    for num in nums:
        freq[num] = freq.get(num, 0) + 1
    freq_list = list(freq.items())
    print(freq_list)
    freq_list.sort(key=lambda x: (x[1], -x[0]), reverse=True)
    print(freq_list)
    result = []
    for i in range(k):
        result.append(freq_list[i][0])
    print(result)

top_k_freq([1,1,1,2,2,3, 3, 4, 4, 5], 4)
```

```python
# Using Hash map and Min Heap
def top_k_freq(nums, k):
    freq = {}
    for num in nums:
        freq[num] = freq.get(num, 0) + 1
    pq = []
    for key, freq in freq.items():
        heapq.heappush(pq, [freq, key])
        print(pq)
        if len(pq)>k:
            heapq.heappop(pq)
    print(pq)
top_k_freq([1, 1, 1, 2, 2, 3, 3, 4, 4, 5], 2)
```

### Pattern 4: Anagrams
Check if Two Strings Are Anagrams:

```python
# two strings are anagram
def is_anagram(s, t):
    return Counter(s) ==  Counter(t)

def is_anagram2(s, t):
    return sorted(s) == sorted(t)

print(is_anagram("listen", "silent"))
print(is_anagram2("listen", "silent"))
```

```python
# Group Anagram
def group_anagram(words):
    groups = {}
    for word in words:
        key = '.'.join(sorted(word))
        groups.setdefault(key, []).append(word)
    return list(groups.values())

print(group_anagram(["eat","tea","tan","ate","nat","bat"]))
```

```python
# Using Tuple of Counts
def group_anagram(words):
    groups = {}
    for word in words:
        count = [0] * 26
        for char in word:
            count[ord(char) - ord('a')] += 1
        key = tuple(count)
        groups.setdefault(key, []).append(word)
    return list(groups.values())
print(group_anagram(["eat","tea","tan","ate","nat","bat"]))
```

### Pattern 5: Subarray Sum Equals K
Use a map to track prefix sums — if prefix[j] - prefix[i] = k, then subarray [i+1..j] sums to k.

```python
def sub_array_sum(arr, k):
    count = 0
    prefix = 0
    seen = {0 : 1}
    for num in arr:
        prefix += num
        complement = abs(k - prefix)
        count += seen.get(complement, 0)
        seen[prefix] = seen.get(prefix, 0) + 1
    return count  
print(sub_array_sum([1, 2, 3], 3))
```

### Pattern 6: Longest Consecutive Sequence
Find the longest sequence of consecutive integers in O(n).

```python
def longest_consecutive(nums):
    sorted_list = sorted(nums)
    print(sorted_list)
    temp = []
    result = []
    for i in sorted_list:
        if len(temp) == 0:
            temp.append(i)
            continue
        if temp[-1]+1 == i:
            temp.append(i)
            print(temp)
        else:
            if len(result) < len(temp):
                result = temp.copy()
                temp.clear()
            temp.append(i)
    return len(result)

longest_consecutive([100,4,200,1,3,2])

# Another way
def longest_consecutive_2(nums):
    num_set = set(nums)
    max_length = 0
    length = 0
    for num in nums:
        if num - 1 in num_set:
            length = 1 
            current = num - 1
            while current  in num_set:
                length += 1
                current = current - 1
            max_length = max(max_length, length)
    return max_length

longest_consecutive_2([100,4,200,1,3,2])
```

#### Pattern 7: Valid Sudoku (Sets for Constraint Checking)


```python
def is_valid_sudoku(board):
    rows = [set() for _ in range(9)]
    cols = [set() for _ in range(9)]
    boxes = [set() for _ in range(9)]
    for r in range(9):
        for c in range(c):
            if board[r][c] == '.':
                continue
            num = board[r][c]
            box_idx = (r//3)*3+(c//3)
            if num in rows[r] or num in cols[c] or num in boxes[box_idx]:
                return False
            rows[r].add(num)
            cols[c].add(num)
            boxes[box_idx].add(num)
    return True
```

```python
### Pattern 8: Intersection and Union

def intersection(nums1: list[int], nums2: list[int]) -> list[int]:
    return list(set(nums1) & set(nums2))

def union(nums1: list[int], nums2: list[int]) -> list[int]:
    return list(set(nums1) | set(nums2))

def difference(nums1: list[int], nums2: list[int]) -> list[int]:
    return list(set(nums1) - set(nums2))

print(intersection([1,2,2,1], [2,2]))
print(union([1,2,2,1], [2,2]))
print(difference([1,2,2,1], [2,2]))
```

```python
# Intersection of Multiple Arrays
def intersection_multiple(arrays):
    if not arrays:
        return []
    result = set(arrays[0])
    for arr in arrays[1:]:
        result &= set(arr)
    return sorted(result)

intersection_multiple([[3,1,2,4,5],[1,2,3,4],[3,4,5,6]])
```

## Pattern 9: Isomorphic Strings

Check if two strings have the same pattern of character mappings.

```python
def is_isomorphic(str1, str2):
    if len(str1) != len(str2):
        return False
    temp = {}
    for i in range(len(str1)):
        if str1[i] in temp.keys():
            if temp[str1[i]] != str2[i]:
                return False
        else:
            temp[str1[i]] = str2[i]
    return True

is_isomorphic("foo", "bar")        
```
