### Recursion

Recursion is when a function calls itself to solve smaller instances of the same problem. Every recursive solution has two parts:
1. Base Case — the stopping condition (prevents infinite recursion)
2. Recursive Case — breaks the problem into smaller subproblems

Each recursive call creates a new stack frame with its own local variables. When the base case is hit, calls unwind back up the stack.

### Pattern 1: Linear Recursion
Process elements one at a time, making a single recursive call.

```python
arr = [1, 2, 3, 4, 5]
def sum_array(arr):
    if len(arr) == 0:
        return 0
    return arr[0] + sum_array(arr[1:])

sum_array(arr)
```

```python
# Reverse String:
def reverse_string(s):
    if len(s) == 0:
        return ""
    return reverse_string(s[1:]) + s[0]

reverse_string("kiran")
```

### Pattern 2: Binary Recursion
Make two recursive calls — splits the problem in half.

```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1)+fibonacci(n-2)

fibonacci(5)
```

```python
# Binary Search (Recursive)
def binary_search(arr, start, end, target):
    if start > end:
        return -1
    mid = start + (end-start) // 2
    if arr[mid] == target:
        return mid
    elif target > arr[mid]:
        return binary_search(arr, mid+1, end, target)
    else:
        return binary_search(arr, start, mid-1, target)

arr = [1, 2, 3, 4, 5, 6]
binary_search(arr, 0, len(arr)-1, 6)
```

### Pattern 3: Multiple Recursion (Tree Recursion)
More than two recursive calls — explores all possibilities.

```python
# Print All Subsequences:
def print_subsequence(s, current):
    if len(s)==0:
        print(current)
        return 
    print_subsequence(s[1:], current + s[0])
    print_subsequence(s[1:], current)
    
print_subsequence("abc", "")
```

### Pattern 4: Divide and Conquer
Divide problem into independent subproblems, solve recursively, combine results.

```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr)//2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    return merge(left, right)

def merge(left_arr, right_arr):
    result = []
    i = j = 0
    while i < len(left_arr) and j < len(right_arr):
        if left_arr[i] > right_arr[j]:
            result.append(right_arr[j])
            j += 1
        else:
            result.append(left_arr[i])
            i += 1
    while i < len(left_arr):
        result.append(left_arr[i])
        i += 1
    while j < len(right_arr):
        result.append(right_arr[j])
        j += 1
    
    return result

merge_sort([5, 4, 3, 2, 1, 0])

```

```python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr)//2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quick_sort(left) + middle + quick_sort(right)

quick_sort([5, 4, 3, 2, 1, 0])
```

#### Pow(x, n)

Implement pow(x, n), which calculates x raised to the power n (i.e., xn).

Example 1:

Input: x = 2.00000, n = 10

Output: 1024.00000

```python
# Recursive approach with divide and concuer
def mypow_recursive(x, n):
    if n == 0:
        return 1
    half = mypow_recursive(x, n // 2)
    if n % 2 == 0:
        return half * half
    return x * half * half

mypow_recursive(2, 10)
```

#### Decode String

The encoding rule is: k[encoded_string], where the encoded_string inside the square brackets is being repeated exactly k times. Note that k is guaranteed to be a positive integer.

Input: s = "3[a]2[bc]"

Output: "aaabcbc"

```python
def decode_string(input):
    result = ""
    i = 0
    stack = []
    while(i < len(input)):
        value = input[i]
        if value == "]":
            temp = ""
            while(stack):
                value1 = stack.pop()
                if value1 == "[":
                    break
                temp += value1
            digit = int(stack.pop())
            stack.append(digit * temp)
        else:
            stack.append(value)
        i += 1
    return "".join(stack)
decode_string("3[a2[c]]")     
```
