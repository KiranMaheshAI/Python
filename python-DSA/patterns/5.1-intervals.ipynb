### Intervals
An interval represents a continuous range with a start and end point. In coding problems, intervals are typically represented as arrays or objects with two values.

## Task Scheduling / Meeting Rooms

Meeting Rooms II — minimum number of rooms needed.

```python
import heapq
intervals = [[0, 30], [15, 20], [5, 10]]
def min_meeting_rooms(intervals):
    if not intervals:
        return 0
    intervals.sort(key=lambda k: k[0])
    print(intervals)
    min_heap = []
    for start, end in intervals:
        if min_heap and min_heap[0] <= start:
            heapq.heapreplace(min_heap, end)
        else:
             heapq.heappush(min_heap, end)
    print(min_heap)
    return len(min_heap)
print(min_meeting_rooms(intervals))
```

### Interval List Intersections 

Given two lists of closed intervals, each list is pairwise disjoint (no overlaps within a list) and sorted. Return the intersection of these two interval lists.

Input: firstList = [[0,2],[5,10],[13,23],[24,25]]

       secondList = [[1,5],[8,12],[15,24],[25,26]]

Output: [[1,2],[5,5],[8,10],[15,23],[24,24],[25,25]]

```python
def interval_list_intersection(first_list, second_list):
    if not first_list or not second_list:
        return []
    result = []
    i, j = 0, 0
    while i<len(first_list) and j < len(second_list):
        a = first_list[i]
        b = second_list[j]
        if a[0] <= b[1] and a[1] >= b[0]:
            result.append([max(a[0], b[0]), max(a[1], b[1])])

        if a[1]< b[1]:
            i+=1
        else:
            j+=1
    return result
```

### Pattern : Activity Selection / Interval Scheduling
Problem: Given intervals with start/end times, select the maximum number of non-overlapping intervals.

Greedy Choice: Always pick the interval that ends earliest — this leaves maximum room for future intervals.

```python
input = [[1,3], [2,4], [3,5], [0,2]]
def max_non_overlapping_intervals(input):
    if not input:
        return []
    input.sort(key=lambda x: x[1])
    print(input)
    result = [input[0]]
    for i in range(1, len(input)):
        temp = input[i]
        if temp[0] >= result[-1][1]:
            result.append(temp)
    print(result)
    return result

max_non_overlapping_intervals(input)
```

```python
input = [[1,3], [2,4], [3,5], [0,6]]
def max_non_overlapping(intervals):
    if not intervals:
        return 0
    # sort by slots end time
    intervals.sort(key = lambda x: x[1])
    print(intervals)
    count = 1
    last_end = intervals[0]
    for interval in intervals[1:]:
        if  interval[0]  >= last_end[1]:
            count += 1
            last_end = interval
    return count

max_non_overlapping([[1,3],[2,4],[3,5],[0,6]])    
```

## Pattern : Minimum Platforms / Meeting Rooms

Given arrival/departure times, find the minimum number of platforms needed.

```python
import heapq

arrivals   = [900, 940, 950, 1100, 1500, 1800]
departures = [910, 1200, 1120, 1130, 1900, 2000]

def min_platforms(arrivals, departures):
    if len(arrivals) != len(departures):
        return 0
    pq = [departures[0]]
    count = 1
    for i in range(1, len(departures)):
        print("Start: ", pq)
        if pq[0] <= arrivals[i]:
            heapq.heappop(pq)
            heapq.heappush(pq, departures[i])
        else:
            count +=1 
            heapq.heappush(pq, departures[i])
        
        print("END: ", pq)
    return count

print(min_platforms(arrivals, departures))
```

## Pattern : Jump Game

**Problem:** Given an array where `nums[i]` is the max jump from position `i`, can you reach the last index?


```python
input = [2, 3, 1, 1, 4]
input2 = [3,2,1,0,4]
def can_jump(input):
    max_reach = 0
    for i in range(len(input)):
        if i > max_reach:
            return False
        jump = i+input[i]
        max_reach = max(max_reach, jump)
        if max_reach >= len(input)-1:
            return True
    return True

can_jump(input2)
```

**Problem:** Find the **minimum number of jumps** to reach the end.

```python
def min_jumps(nums: list[int]) -> int:
    if len(nums) <= 1:
        return 0
    jumps = 0
    current_end = 0        # end of current jump range
    farthest = 0           # farthest reachable in next jump
    for i in range(len(nums) - 1):
        farthest = max(farthest, i + nums[i])
        if i == current_end:       # reached end of current jump
            jumps += 1
            current_end = farthest
            if current_end >= len(nums) - 1:
                break
    return jumps

min_jumps([2,3,1,1,4])
```

## Pattern 5: Non-overlapping Intervals

**Problem:** Find the **minimum number of intervals to remove** to make the rest non-overlapping.

```python
input = [[1,2],[2,3],[3,4],[1,3]]
def erase_overlap_intervals(intervals):
    if not intervals:
        return 0
    intervals.sort(key=lambda x: x[1])
    count = 0
    last_end = input[0][1]
    for start, end in intervals[1:]:
        if start >= last_end:
            last_end = end
        else:
            count += 1
    return count

erase_overlap_intervals(input)
```

## Common Mistakes

1. Forgetting to sort

2. Using Wrong sort key: Sorting by start time and end time are not interchangeable. Using the wrong one gives incorrect results.

    a. Merge intervals: Sort by start time

    b. Select maximum non-overlapping: Sort by end time
    
    c. Insert interval: Input is already sorted by start time

3. Off-by-One in Overlap Detection : The difference between < and <= matters for whether touching intervals count as overlapping.

4. Forgetting the Last Interval : When building a result list incrementally, the last interval is often added inside the loop only under certain conditions. Make sure to add it after the loop.

5. Modifying Input Array : Some problems expect the original array to remain unchanged. Creating a copy before sorting avoids this issue.
