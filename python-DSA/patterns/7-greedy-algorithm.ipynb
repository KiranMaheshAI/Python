## Greedy Algorithm

A greedy algorithm makes the locally optimal choice at each step, hoping to find a global optimum. It never reconsiders earlier decisions — once a choice is made, it's final.

Greedy vs Dynamic Programming:

Greedy -> Make the best choice NOW, never look back

DP ->  Consider all possible choices, pick the best overall

How to Recognize Greedy Problems

Look for these signals:
1. Optimization words: minimize, maximize, largest, smallest, most, least
2. Selection problems: choose items, schedule tasks, assign resources
3. Ordering matters: sorting often helps reveal the greedy choice
4. Exchange argument: swapping any two choices makes things worse
5. No dependencies: future choices don't invalidate past ones

### Pattern 1: Activity Selection / Interval Scheduling
Problem: Given intervals with start/end times, select the maximum number of non-overlapping intervals.

Greedy Choice: Always pick the interval that ends earliest — this leaves maximum room for future intervals.

```python
input = [[1,3], [2,4], [3,5], [0,2]]
def max_non_overlapping_intervals(input):
    if not input:
        return []
    input.sort(key=lambda x: x[1])
    print(input)
    result = [input[0]]
    for i in range(1, len(input)):
        temp = input[i]
        if temp[0] >= result[-1][1]:
            result.append(temp)
    print(result)
    return result

max_non_overlapping_intervals(input)
```

```python
input = [[1,3], [2,4], [3,5], [0,6]]
def max_non_overlapping(intervals):
    if not intervals:
        return 0
    # sort by slots end time
    intervals.sort(key = lambda x: x[1])
    print(intervals)
    count = 1
    last_end = intervals[0]
    for interval in intervals[1:]:
        if  interval[0]  >= last_end[1]:
            count += 1
            last_end = interval
    return count

max_non_overlapping([[1,3],[2,4],[3,5],[0,6]])    
```

## Pattern 2: Minimum Platforms / Meeting Rooms

Given arrival/departure times, find the minimum number of platforms needed.

```python
import heapq

arrivals   = [900, 940, 950, 1100, 1500, 1800]
departures = [910, 1200, 1120, 1130, 1900, 2000]

def min_platforms(arrivals, departures):
    if len(arrivals) != len(departures):
        return 0
    pq = [departures[0]]
    count = 1
    for i in range(1, len(departures)):
        print("Start: ", pq)
        if pq[0] <= arrivals[i]:
            heapq.heappop(pq)
            heapq.heappush(pq, departures[i])
        else:
            count +=1 
            heapq.heappush(pq, departures[i])
        
        print("END: ", pq)
    return count

print(min_platforms(arrivals, departures))
```

## Pattern 4: Jump Game

**Problem:** Given an array where `nums[i]` is the max jump from position `i`, can you reach the last index?


```python
input = [2, 3, 1, 1, 4]
input2 = [3,2,1,0,4]
def can_jump(input):
    max_reach = 0
    for i in range(len(input)):
        if i > max_reach:
            return False
        jump = i+input[i]
        max_reach = max(max_reach, jump)
        if max_reach >= len(input)-1:
            return True
    return True

can_jump(input2)
```

**Problem:** Find the **minimum number of jumps** to reach the end.

```python
def min_jumps(nums: list[int]) -> int:
    if len(nums) <= 1:
        return 0
    jumps = 0
    current_end = 0        # end of current jump range
    farthest = 0           # farthest reachable in next jump
    for i in range(len(nums) - 1):
        farthest = max(farthest, i + nums[i])
        if i == current_end:       # reached end of current jump
            jumps += 1
            current_end = farthest
            if current_end >= len(nums) - 1:
                break
    return jumps

min_jumps([2,3,1,1,4])
```

## Pattern 5: Non-overlapping Intervals

**Problem:** Find the **minimum number of intervals to remove** to make the rest non-overlapping.

```python
input = [[1,2],[2,3],[3,4],[1,3]]
def erase_overlap_intervals(intervals):
    if not intervals:
        return 0
    intervals.sort(key=lambda x: x[1])
    count = 0
    last_end = input[0][1]
    for start, end in intervals[1:]:
        if start >= last_end:
            last_end = end
        else:
            count += 1
    return count

erase_overlap_intervals(input)
```
