```python
# Approach 2 
def ArrayChallenge(arr):
    stack = []
    max_area = 0
    index = 0
    
    # We loop through all bars, plus one extra '0' height bar 
    # at the end to ensure we pop everything remaining in the stack.
    temp_arr = arr + [0]
    
    while index < len(temp_arr):
        # If stack is empty or current bar is taller than stack top, push index
        if not stack or temp_arr[index] >= temp_arr[stack[-1]]:
            stack.append(index)
            index += 1
        else:
            print(f"Stack is {stack} and max_area is {max_area}")
            # Pop the top index to calculate area with that height
            height = temp_arr[stack.pop()]
            
            # If stack is empty, width is the current index
            # Otherwise, width is the distance between current index and new stack top
            width = index if not stack else index - stack[-1] - 1
            
            max_area = max(max_area, height * width)
            
    return max_area

# Test Cases
# print(ArrayChallenge([6, 3, 1, 4, 12, 4])) # Output: 12
print(ArrayChallenge([5, 6, 7, 4, 1]))     # Output: 16
# print(ArrayChallenge([2, 1, 3, 4, 1]))     # Output: 6
    
```
