### Construct a tree from Inorder and Preorder

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

def buildTree(inorder, preorder):
    if not inorder or not preorder:
        return None
    
    root_val = preorder[0]
    root = Node(root_val)
    
    root_index = inorder.index(root_val)
    
    root.left = buildTree(inorder[:root_index], preorder[1:root_index+1])
    root.right = buildTree(inorder[root_index+1:], preorder[root_index+1:])
    
    return root

inorder = [4, 2, 5, 1, 3, 6]
preorder = [1, 2, 4, 5, 3, 6]

root = buildTree(inorder, preorder)

def levelOrder(root):
    if not root:
        return []
    
    queue = [root]
    result = []
    
    while queue:
        node = queue.pop(0)
        result.append(node.data)
        
        if node.left:
            queue.append(node.left)
            
        if node.right:
            queue.append(node.right)
            
    return result

print(levelOrder(root))
    
```

```python
# Another way to construct a tree from Inorder and Preorder
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
        
def buildTree(inorder, preorder, inStart, inEnd, preStart, preEnd):
    if inStart > inEnd or preStart > preEnd:
        return None
    
    root_val = preorder[preStart]
    root = TreeNode(root_val)
    
    root_index = inorder.index(root_val)
    
    left_subtree_size = root_index - inStart
    right_subtree_size = inEnd - root_index
    
    root.left = buildTree(inorder, preorder, inStart, root_index-1, preStart+1, preStart+left_subtree_size)
    root.right = buildTree(inorder, preorder, root_index+1, inEnd, preStart+left_subtree_size+1, preEnd)
    
    return root

inorder = [4, 2, 5, 1, 3, 6]
preorder = [1, 2, 4, 5, 3, 6]
```

## Construct a Binary Tree from Inorder and PostOrder

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
        
def buildTree(inorder, postorder):
    if not inorder or not postorder:
        return None
    
    root_val = postorder[-1]
    root = TreeNode(root_val)
    
    root_index = inorder.index(root_val)

    root.right = buildTree(inorder[root_index+1:], postorder[root_index:-1])
    root.left = buildTree(inorder[:root_index], postorder[:root_index])
    
    return root

inorder = [4, 2, 5, 1, 3, 6]
postorder = [4, 5, 2, 6, 3, 1]
root = buildTree(inorder, postorder)

def levelOrder(root):
    if not root:
        return []
    
    queue = [root]
    result = []
        
    while queue:
        node = queue.pop(0)
        result.append(node.val)
        
        if node.left:
            queue.append(node.left)
            
        if node.right:
            queue.append(node.right)
            
    return result

levelOrder(root)
```

```python
# Another way to construct a tree from Inorder and PostOrder
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
        
def buildTree(inorder, postorder, inStart, inEnd, postStart, postEnd):
    if inStart > inEnd or postStart > postEnd:
        return None
    
    root_val = postorder[postEnd]
    root = TreeNode(root_val)
    
    root_index = inorder.index(root_val)
    
    left_subtree_size = root_index - inStart
    right_subtree_size = inEnd - root_index
    
    root.left = buildTree(inorder, postorder, inStart, root_index-1, postStart, postStart+left_subtree_size-1)
    root.right = buildTree(inorder, postorder, root_index+1, inEnd, postStart+left_subtree_size, postEnd-1)
    
    return root

inorder = [4, 2, 5, 1, 3, 6]
postorder = [4, 5, 2, 6, 3, 1]

root = buildTree(inorder, postorder, 0, 5, 0, 5)

def levelOrder(root):
    if not root:
        return []
    
    queue = [root]
    result = []
    
    while queue:
        node = queue.pop(0)
        result.append(node.val)
        
        if node.left:
            queue.append(node.left)
            
        if node.right:
            queue.append(node.right)
        
    return result

levelOrder(root)
            
```
