```python
class BinaryTreeNode:
    def __init__(self, d):
        self.data = d
        self.left = None
        self.right = None

# Initialize and allocate memory for tree nodes
root = BinaryTreeNode(2)
root.left = BinaryTreeNode(3)
root.right = BinaryTreeNode(4)
root.left.left = BinaryTreeNode(5)
root.left.right = BinaryTreeNode(6)
root.right.left = BinaryTreeNode(7)
root.right.right = BinaryTreeNode(8)
```

```python
# print the binary tree
def print_binary_tree(root):
    if root is None:
        return
    print(root.data, end=" ")
    if root.left is not None:
        print("L -> ", root.left.data, end=" ")
    else:
        print("L -> None", end=" ")
    if root.right is not None:
        print("R -> ", root.right.data, end=" ")
    else:
        print("R -> None", end=" ")
    print()
    print_binary_tree(root.left)
    print_binary_tree(root.right)
print_binary_tree(root)
```

```python
# Take input
def take_tree_input():
    root_data = int(input("Enter the data of the node: "))
    if root_data == -1:
        return None
    root = BinaryTreeNode(root_data)
    print("Enter the left child of ", root_data)
    root.left = take_tree_input()
    print("Enter the right child of ", root_data)
    root.right = take_tree_input()
    return root
root = take_tree_input()
print_binary_tree(root)
```

```python
# Take input level wise
from collections import deque

def take_tree_input_level_wise():
    root_data = int(input("Enter the data of the node: "))
    if root_data == -1:
        return None
    root = BinaryTreeNode(root_data)
    q = deque([root])
    while q:
        current_node = q.popleft()
        left_child_data = int(input("Enter the left child of ", current_node.data, ": "))
        if left_child_data != -1:
            left_child = BinaryTreeNode(left_child_data)
            current_node.left = left_child
            q.append(left_child)
        right_child_data = int(input("Enter the right child of ", current_node.data, ": "))
        if right_child_data != -1:
            right_child = BinaryTreeNode(right_child_data)
            current_node.right = right_child
            q.append(right_child)
    return root 
root = take_tree_input_level_wise()
```

```python
# print the binary tree level wise
from collections import deque

def print_binary_tree_level_wise(root):
    if root is None:
        return
    q = deque([root])
    while q:
        current_node = q.popleft()
        print(current_node.data, end=": ")
        if current_node.left is not None:
            print("L -> ", current_node.left.data, end=", ")
            q.append(current_node.left)
        else:
            print("L -> None, ", end="")
        if current_node.right is not None:
            print("R -> ", current_node.right.data, end="\n")
            q.append(current_node.right)
        else:
            print("R -> None\n")
        
print_binary_tree_level_wise(root)
    
```

```python
# height of the binary tree
def height_of_tree(root):
    if root is None:
        return 0
    return 1 + max(height_of_tree(root.left), height_of_tree(root.right))
print(height_of_tree(root))
```

```python
# diameter of the binary tree where the longest path between any two nodes for the balanced tree
def diameter_of_tree(root):
    if root is None:
        return 0
    return 1 + height_of_tree(root.left) + height_of_tree(root.right)
print(diameter_of_tree(root))


# diameter of the binary tree where the longest path between any two nodes for the unbalanced tree
def diameter_of_tree(root):
    if root is None:
        return 0
    leftHeight = height_of_tree(root.left)
    rightHeight = height_of_tree(root.right)
    leftDiameter = diameter_of_tree(root.left)
    rightDiameter = diameter_of_tree(root.right)
    return max(leftHeight + rightHeight, leftDiameter, rightDiameter)

print(diameter_of_tree(root))
```

```python
# Diameter of the binary tree where the longest path between any two nodes for the unbalanced tree

def diameter_of_tree_optimized(root):
    if root is None:
        return 0, 0 # height, diameter
    leftHeight, leftDiameter = diameter_of_tree_optimized(root.left)
    rightHeight, rightDiameter = diameter_of_tree_optimized(root.right)
    height = 1 + max(leftHeight, rightHeight)
    diameter = max(leftHeight + rightHeight, leftDiameter, rightDiameter)
    return height, diameter
print(diameter_of_tree_optimized(root))
```

```python
# Check if a binary tree is balanced
def is_balanced(root):
    if root is None:
        return True, 0 # isBalanced, height
    isLeftBalanced, leftHeight = is_balanced(root.left)
    isRightBalanced, rightHeight = is_balanced(root.right)
    if not isLeftBalanced or not isRightBalanced:
        return False, 0
    if abs(leftHeight - rightHeight) > 1:
        return False, 0
    return True, 1 + max(leftHeight, rightHeight)
print(is_balanced(root))
```
