## Traversals in Trees:

Traversal means how do we traverse the trees, there are different types of traversals:
1. Pre order traversal
2. In order traversal
3. Post order traversal
4. Level order traversal
5. Zig zag level order traversal

````python
# Pre order traversal:
def pre_order_traversal(root):
    if root is None:
        return
    print(root.get_data(), end=" ")
    for child in root.get_children():
        pre_order_traversal(child)
````

````python
# post order traversal:
def post_order_traversal(root):
    if root is None:
        return
    for child in root.get_children():
        post_order_traversal(child)
    print(root.get_data(), end=" ")
````

## In-Order Traversal (Binary Trees)

**In-order traversal** is specifically for **Binary Trees** where each node has at most 2 children (left and right).

**Order of traversal:**
1. Traverse the **left subtree** (recursively)
2. Visit the **root node**
3. Traverse the **right subtree** (recursively)

**Key Property:** For a Binary Search Tree (BST), in-order traversal gives nodes in **ascending order**.

**Example:**
```
        1
       / \
      2   3
     / \
    4   5
```

In-order traversal: **4 → 2 → 5 → 1 → 3**


````python
# First, let's create a Binary Tree Node class
class BinaryTreeNode:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None
````

````python
# In-order traversal for Binary Trees
def in_order_traversal(root):
    """
    In-order: Left -> Root -> Right
    """
    if root is None:
        return
    
    # First, traverse left subtree
    in_order_traversal(root.left)
    
    # Then, visit the root
    print(root.data, end=" ")
    
    # Finally, traverse right subtree
    in_order_traversal(root.right)
````

````python
# Example 1: Simple Binary Tree
"""
Tree Structure:
        1
       / \
      2   3
     / \
    4   5
"""

# Create the tree
root = BinaryTreeNode(1)
root.left = BinaryTreeNode(2)
root.right = BinaryTreeNode(3)
root.left.left = BinaryTreeNode(4)
root.left.right = BinaryTreeNode(5)

print("In-order Traversal (Left -> Root -> Right):")
in_order_traversal(root)
print()

# Expected Output: 4 2 5 1 3
````

````python
# Example 2: Binary Search Tree (BST)
"""
BST Structure:
        50
       /  \
      30   70
     / \   / \
   20  40 60  80
"""

# Create a BST
bst_root = BinaryTreeNode(50)
bst_root.left = BinaryTreeNode(30)
bst_root.right = BinaryTreeNode(70)
bst_root.left.left = BinaryTreeNode(20)
bst_root.left.right = BinaryTreeNode(40)
bst_root.right.left = BinaryTreeNode(60)
bst_root.right.right = BinaryTreeNode(80)

print("In-order Traversal of BST (gives sorted order):")
in_order_traversal(bst_root)
print()

# Expected Output: 20 30 40 50 60 70 80 (sorted!)
````

````python
# In-order traversal (Iterative approach using stack)
def in_order_traversal_iterative(root):
    """
    In-order traversal using stack (iterative)
    """
    if root is None:
        return
    
    stack = []
    current = root
    
    while current or stack:
        # Go to the leftmost node
        while current:
            stack.append(current)
            current = current.left
        
        # Current is None, pop from stack
        current = stack.pop()
        print(current.data, end=" ")
        
        # Now visit right subtree
        current = current.right

print("In-order Traversal (Iterative):")
in_order_traversal_iterative(root)
print()
````

## Comparison of All Three Traversals

Using the same tree:
```
        1
       / \
      2   3
     / \
    4   5
```

- **Pre-order** (Root → Left → Right): **1 → 2 → 4 → 5 → 3**
- **In-order** (Left → Root → Right): **4 → 2 → 5 → 1 → 3**
- **Post-order** (Left → Right → Root): **4 → 5 → 2 → 3 → 1**


````python
# Compare all three traversals
print("Pre-order (Root -> Left -> Right):", end=" ")
def pre_order(root):
    if root:
        print(root.data, end=" ")
        pre_order(root.left)
        pre_order(root.right)

pre_order(root)
print()

print("In-order (Left -> Root -> Right):", end=" ")
in_order_traversal(root)
print()

print("Post-order (Left -> Right -> Root):", end=" ")
def post_order(root):
    if root:
        post_order(root.left)
        post_order(root.right)
        print(root.data, end=" ")

post_order(root)
print()
````

## Level Order Traversal (Breadth-First Search)

**Level order traversal** visits all nodes at each level before moving to the next level. It goes **left to right** at each level.

**Algorithm:** Uses a **Queue** data structure (FIFO - First In First Out)

**Steps:**
1. Start with the root node in the queue
2. While queue is not empty:
   - Dequeue a node
   - Visit/print that node
   - Enqueue all its children (left to right)

**Example:**
```
        1
       / \
      2   3
     / \
    4   5
```

Level order traversal: **1 → 2 → 3 → 4 → 5**

**Levels:**
- Level 0: 1
- Level 1: 2, 3
- Level 2: 4, 5


````python
from collections import deque

# Level order traversal for Binary Trees
def level_order_traversal(root):
    """
    Level order traversal using queue (BFS)
    """
    if root is None:
        return
    
    # Create a queue and add root
    queue = deque([root])
    
    while queue:
        # Dequeue a node
        current = queue.popleft()
        
        # Print the node
        print(current.data, end=" ")
        
        # Enqueue left child
        if current.left:
            queue.append(current.left)
        
        # Enqueue right child
        if current.right:
            queue.append(current.right)
````

````python
# Example 1: Level order traversal on our simple tree
"""
Tree Structure:
        1
       / \
      2   3
     / \
    4   5
"""

print("Level Order Traversal:")
level_order_traversal(root)
print()

# Expected Output: 1 2 3 4 5
````

## Applications of Level Order Traversal

1. **Finding height/depth of tree** - count number of levels
2. **Finding nodes at a specific level**
3. **Checking if tree is complete binary tree**
4. **Printing tree level by level**
5. **Serialization/Deserialization of trees**
6. **Finding leftmost/rightmost node at each level**


````python
# Application 2: Get nodes at a specific level
def get_nodes_at_level(root, target_level):
    """
    Get all nodes at a specific level (0-indexed)
    """
    if root is None:
        return []
    
    queue = deque([root])
    current_level = 0
    
    while queue:
        if current_level == target_level:
            # Return all nodes at this level
            return [node.data for node in queue]
        
        level_size = len(queue)
        for i in range(level_size):
            node = queue.popleft()
            
            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)
        
        current_level += 1
    
    return []

print(f"Nodes at level 0: {get_nodes_at_level(root, 0)}")
print(f"Nodes at level 1: {get_nodes_at_level(root, 1)}")
print(f"Nodes at level 2: {get_nodes_at_level(root, 2)}")
````

## Summary: All Traversals Comparison

**DFS (Depth First Search) - Uses Stack/Recursion:**
- **Pre-order**: Root → Left → Right
- **In-order**: Left → Root → Right (gives sorted order for BST)
- **Post-order**: Left → Right → Root

**BFS (Breadth First Search) - Uses Queue:**
- **Level-order**: Level by level, left to right

**Time Complexity:** O(n) for all traversals (visit each node once)

**Space Complexity:**
- Recursive DFS: O(h) where h is height (call stack)
- Level Order: O(w) where w is maximum width of tree
