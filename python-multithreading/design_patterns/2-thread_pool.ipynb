## Thread Pool Pattern

In this pattern, there is a pool of worker threads that can be reused for executing tasks. Using a pool removes the overhead of creating and destroying threads. Great for executing a large number of short-lived tasks.

The Thread Pool Pattern is a concurrency design pattern that pre-initializes a set of worker threads to execute tasks from a queue, enhancing performance by reducing the overhead of creating and destroying threads for short-lived tasks. It limits resource consumption, manages concurrent execution, and improves system stability

#### Using ThreadPoolExecutor
The most recommended way to use a thread pool is via the context manager (with statement), which ensures the pool is shut down properly and its resources are released automatically. 

Key Methods
1. submit(fn, *args, **kwargs): Schedules a function (fn) to be executed in the pool with the given arguments. It returns a Future object, which represents the result of the task that will be available later.
2. map(func, iterable, timeout=None): A parallel equivalent of the built-in map() function. It applies a function to each item in an iterable, submitting them as tasks to the pool and returning the results in the order the tasks were submitted.
3. shutdown(wait=True): Releases the resources associated with the executor. If wait is True (default), it blocks until all submitted tasks are completed. Using a with statement calls this automatically. 

```python
import concurrent.futures
import time

def task(name):
    """A sample I/O-bound task that sleeps for a bit."""
    print(f"Task {name}: starting")
    time.sleep(2) # Simulate I/O work (e.g., network request, file I/O)
    print(f"Task {name}: finished")
    return f"Result {name}"

# Use a context manager to create and manage the thread pool
with concurrent.futures.ThreadPoolExecutor(max_workers=3) as executor:
    # Submit tasks to the pool and get Future objects
    future1 = executor.submit(task, "A")
    future2 = executor.submit(task, "B")
    future3 = executor.submit(task, "C")
    
    # Retrieve results from the Future objects (this blocks until the task is done)
    print(f"\n{future1.result()}")
    print(f"{future2.result()}")
    print(f"{future3.result()}\n")

print("All tasks are done and the thread pool is shut down.")
```
