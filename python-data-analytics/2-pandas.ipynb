#### Pandas-DataFrame And Series
Pandas is a powerful data manipulation library in Python, widely used for data analysis and data cleaning. It provides two primary data structures: Series and DataFrame. A Series is a one-dimensional array-like object, homogeneous array and size immutable, while a DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns).

Why pandas:
1. import data from different formats liks csv, excel.
2. Data Cleaning: Used for detecting missing values and invalid values.
3. Size mutability: We can add or delete from a dataset
4. Reshaping and pivoting of datasets.
5. Effective manipulation and extraction.
6. Statistical analysis

```python
## Create a Series from dictionary
data={'a':1,'b':2,'c':3}
series_dict=pd.Series(data)
print(series_dict)
```

```python
# Logical selection
series_dict[(series_dict>=2)&(series_dict<=3)]
```

```python
## Data Cleaning
df.isnull().sum()
```

```python
# Drop null values rows
# Use dropna for getting rid of them
df.dropna(how="any") # any row value is null
df.dropna(how="all") # all row value is null
df.dropna(subset=["Category_New"]) # only Category_New column value is null
df
df.isnull().sum()
```

```python
# Use fillna for replacing the null values
df.fillna(0)
```

```python
df["Sales"].fillna(df["Sales"].mean())
```

```python
# Instead of ffilling with the default values, we can use forward fill and backward fill
# Forward fill means that the null value will be replaced with the previous value
df.fillna(method="ffill")
# Backward fill means that the null value will be replaced with the next value
df.fillna(method="bfill")

# Problem with the Forward fill is if the first value is null, then next values also will be null. 
```

```python
# Duplicates
df_duplicates = df[df.duplicated()]
df_duplicates
# if you want the first duplicated values and remove the next one
df_duplicates = df[df.duplicated(keep='first')]
df_duplicates
# if you want the last duplicated values and remove the first one
df_duplicates = df[df.duplicated(keep='last')]
df_duplicates
```

```python
# invalid values
df["Updated Value"] = df["Value"].apply(lambda x: x*2 if x > 50.0 else x) 
df
```
