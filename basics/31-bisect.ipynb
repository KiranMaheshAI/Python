## Bisect 

In Python, the bisect module provides support for maintaining a list in sorted order without having to sort the list after every insertion. It uses a binary search algorithm to find the correct insertion point for an element.

### Core Functions:

The module offers two main types of functions: those that find an index and those that perform the insertion:

1. bisect_left(a, x, lo=0, hi=len(a), *, key=None): Returns the leftmost insertion point for x in sorted list a. If x is already present, the insertion point will be before any existing entries.
2. bisect_right(a, x, lo=0, hi=len(a), *, key=None): Returns the rightmost insertion point for x in sorted list a. If x is already present, the insertion point will be after any existing entries.
3. bisect: An alias for bisect_right.
4. insort_left(a, x, ...): Inserts x into a at the position returned by bisect_left.
5. insort_right(a, x, ...): Inserts x into a at the position returned by bisect_right.
6. insort: An alias for insort_right

```python
import bisect
li = [1, 3, 4, 4, 4, 6, 7] 

print(bisect.bisect(li, 4)) # right
print(bisect.bisect_left(li, 4)) # left
print(bisect.bisect_right(li, 4, 0, 4)) # subright
```

```python
import bisect

l1 = [1, 3, 4, 4, 4, 6, 7]  
l2 = [1, 3, 4, 4, 4, 6, 7]  
l3 = [1, 3, 4, 4, 4, 6, 7]  

bisect.insort(l1, 5) # right
print(l1) 

bisect.insort_left(l2, 5) # left
print(l2)

bisect.insort_right(l3, 5, 0, 4) # subright
print(l3)
```
