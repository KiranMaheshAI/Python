## Abstract Base Class

The concept of interfaces can be effectively implemented and enforced using Abstract Base Classes (ABCs) from the abc module. Interfaces focus on defining a contract between a class and its users, while abstract classes focus on defining a common interface for a group of related classes.

### Using Abstract Base Classes (ABCs) for Interfaces:
This is the most common way to create interfaces in Python. An interface, in this context, is an abstract class where all methods are declared as abstract methods.

#### Defining an interface:

-> Import `ABC` and `abstractmethod` from the abc module.

-> Create a class that inherits from `ABC`.

-> Decorate all methods within this class with `@abstractmethod`. These methods will have no implementation in the interface itself.

#### Implementing an interface:

-> Create a class that inherits from the defined interface.

-> This implementing class must provide concrete implementations for all abstract methods declared in the interface.

-> Attempting to instantiate a class that inherits from an ABC without implementing all abstract methods will result in a `TypeError`.

```python
from abc import ABC, abstractmethod

class Shape(ABC): # Define the interface
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

class Circle(Shape): # Implement the interface
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

    def perimeter(self):
        return 2 * 3.14 * self.radius

class Square(Shape): # Implement the interface
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side * self.side

    def perimeter(self):
        return 4 * self.side

# Usage
circle = Circle(5)
print(f"Circle Area: {circle.area()}")

square = Square(4)
print(f"Square Perimeter: {square.perimeter()}")
```

```python
from interface import implements, Interface

class MyInterface(Interface):

    def method1(self, x):
        pass

    def method2(self, x, y):
        pass


class MyClass(implements(MyInterface)):

    def method1(self, x):
        return x * 2

    def method2(self, x, y):
        return x + y

# Usage
my_object = MyClass()
print(my_object.method1(5))
print(my_object.method2(3, 4))
```

## Zope.interface

In object-oriented languages like Python, the interface is a collection of method signatures that should be provided by the implementing class. Implementing an interface is a way of writing an organized code and achieve abstraction. The package zope.interface provides an implementation of "object interfaces" for Python. It is maintained by the Zope Toolkit project.

```python
import zope.interface


class MyInterface(zope.interface.Interface):
    x = zope.interface.Attribute('foo')
    def method1(self, x, y, z):
        pass
    def method2(self):
        pass

@zope.interface.implementer(MyInterface)
class MyClass:
    def method1(self, x):
        return x**2
    def method2(self):
        return "foo"
obj = MyClass()

# ask an interface whether it 
# is implemented by a class:
print(MyInterface.implementedBy(MyClass))

# MyClass does not provide 
# MyInterface but implements it:
print(MyInterface.providedBy(MyClass))

# ask whether an interface
# is provided by an object:
print(MyInterface.providedBy(obj))

# ask what interfaces are 
# implemented by a class:
print(list(zope.interface.implementedBy(MyClass)))

# ask what interfaces are
# provided by an object:
print(list(zope.interface.providedBy(obj)))

# class does not provide interface
print(list(zope.interface.providedBy(MyClass)))
```
